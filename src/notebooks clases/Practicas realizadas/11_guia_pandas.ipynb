Realizá las siguientes consultas:

### <span style="color:yellow">**Ejercicio 7** — Merge: combinar DataFrames</span>

Tenés dos DataFrames separados:

```python
df_empleados = pd.DataFrame({
    'id':      [1, 2, 3, 4, 5],
    'nombre':  ['Ana', 'Bruno', 'Carla', 'Diego', 'Elena'],
    'area_id': [1, 2, 1, 3, 2],
})

df_areas = pd.DataFrame({
    'area_id': [1, 2, 3, 4],
    'area':    ['Tecnología', 'Marketing', 'Finanzas', 'RRHH'],
    'presupuesto': [500000, 300000, 400000, 200000],
})

df_salarios = pd.DataFrame({
    'id':      [1, 2, 3, 5],   # Diego (id=4) no tiene salario registrado
    'salario': [95000, 78000, 102000, 71000],
})
```

1. Hacé un `inner join` entre empleados y áreas por `area_id`
2. Hacé un `left join` entre empleados y salarios por `id` (conserva a Diego)
3. Combiná los tres en un solo DataFrame
4. Mostrá qué empleados no tienen salario registrado

````python
import pandas as pd

df_empleados: pd.DataFrame = pd.DataFrame({
    'id':      [1, 2, 3, 4, 5],
    'nombre':  ['Ana', 'Bruno', 'Carla', 'Diego', 'Elena'],
    'area_id': [1, 2, 1, 3, 2],
})
df_areas: pd.DataFrame = pd.DataFrame({
    'area_id':     [1, 2, 3, 4],
    'area':        ['Tecnología', 'Marketing', 'Finanzas', 'RRHH'],
    'presupuesto': [500000, 300000, 400000, 200000],
})
df_salarios: pd.DataFrame = pd.DataFrame({
    'id':      [1, 2, 3, 5],
    'salario': [95000, 78000, 102000, 71000],
})


# ==========================================
# RESOLUCIÓN DE CONSULTAS (Esta parte sí estaba bien lógica)
# ==========================================


# 1. Inner join entre empleados y áreas por 'area_id'
join_1 = pd.merge(df_empleados, df_areas, on='area_id', how='inner')
#print(join_1)

# 2. Left join entre empleados y salarios por 'id' (conserva a Diego)
join_2 = pd.merge(df_empleados, df_salarios, on='id', how='left_anti')
#print(join_2)

# 3. Combinación de los tres en un solo DataFrame
df_completo = pd.merge(df_empleados, df_areas, on='area_id', how='left')
df_completo = pd.merge(df_completo, df_salarios, on='id', how='left')
#print(df_completo)

# 4. Empleados que no tienen salario registrado
sin_salario = df_completo[df_completo['salario'].isna()]
print(sin_salario)
````

### <span style="color:yellow">**Ejercicio 8** — Leer CSV y análisis exploratorio</span>

Generá un CSV de ventas, guardálo, leélo con Pandas y hacé un análisis exploratorio completo:

```python
# Primero: generar y guardar el CSV
np.random.seed(42)
n = 200
df_gen = pd.DataFrame({
    'fecha':    pd.date_range('2024-01-01', periods=n, freq='D').strftime('%Y-%m-%d'),
    'vendedor': np.random.choice(['Ana','Bruno','Carla','Diego','Elena'], n),
    'producto': np.random.choice(['Laptop','Mouse','Monitor','Teclado','Auriculares'], n),
    'cantidad': np.random.randint(1, 8, n),
    'precio':   np.random.choice([1500, 25, 800, 120, 350], n),
})
df_gen['total'] = df_gen['cantidad'] * df_gen['precio']
df_gen.to_csv('ventas_2024.csv', index=False)
```

Luego leélo y respondé:
1. ¿Cuántas filas y columnas tiene?
2. ¿Hay valores nulos?
3. ¿Cuál es el total facturado?
4. ¿Cuál es el producto más vendido por cantidad total?
5. ¿Cuál es el vendedor con mayor facturación?

````python
# Primero: generar y guardar el CSV
np.random.seed(42)
n = 200
df_gen = pd.DataFrame({
    'fecha':    pd.date_range('2024-01-01', periods=n, freq='D').strftime('%Y-%m-%d'),
    
    'producto': np.random.choice(['Laptop','Mouse','Monitor','Teclado','Auriculares'], n),
    'vendedor': np.random.choice(['Ana','Bruno','Carla','Diego','Elena'], n),
    'cantidad': np.random.randint(1, 8, n),
    'precio':   np.random.choice([1500, 25, 800, 120, 350], n),
})
df_gen['total'] = df_gen['cantidad'] * df_gen['precio']

df_gen.to_csv('ventas_2024.csv', index=False)

# Leer el archivo
df: pd.DataFrame = pd.read_csv('ventas_2024.csv')

#1 dimensiones
print("vendedores: ",df.shape[0], " columnas: ", df.shape[1])

#2 Valores nulos
print(f" Valores nulos: {df.isna().sum()}")

#3 Total facturado
print(f"Total Facturado: {df["total"].sum():,.0f}")

#4 Producto más vendido por cantidad
mas_vendido = df.groupby('producto')['cantidad'].sum().idxmax()
cantidad_mas_vendido = df.groupby('producto')['cantidad'].sum().max()
print(f"Producto más vendido: {mas_vendido}, con: {cantidad_mas_vendido}")

#5 Vendedor con mayor facturación
mas_vendedor = df.groupby('precio')['total'].sum().idxmax()
monto_mas_vendedor = df.groupby('precio')['total'].sum().max()
print(f"Vendedor con más ventas: {mas_vendedor}, con: {monto_mas_vendedor:,.0f}")
````
