### **Ejercicio 2** — Gráfico de barras: ventas por vendedor

Creá un gráfico de barras verticales con las ventas anuales por vendedor:

```python
vendedores = ['Ana', 'Bruno', 'Carla', 'Diego', 'Elena']
ventas     = [320000, 285000, 410000, 195000, 375000]
```

Requisitos:
- Barras de distintos colores
- Mostrar el valor encima de cada barra
- Título y etiquetas de ejes
- Grilla solo en el eje Y
- Resaltá la barra del vendedor con mayor venta usando un color diferente

````python
import matplotlib.pyplot as plt

vendedores: list[str] = ['Ana', 'Bruno', 'Carla', 'Diego', 'Elena']
ventas:     list[int] = [320000, 285000, 410000, 195000, 375000]

idx_mejor: int = ventas.index(max(ventas))
colores: list[str] = ['steelblue' if i != idx_mejor else 'gold' for i in range(len(ventas))]

fig, ax = plt.subplots(figsize=(8, 5))

barras = ax.bar(vendedores, ventas, color=colores, edgecolor='white')

for barra in barras:
    alto: float = barra.get_height()
    ax.text(barra.get_x() + barra.get_width() / 2, alto + 5000,
            f'${alto:,.0f}', ha='center', va='bottom', fontsize=9)

ax.set_title('Ventas Anuales por Vendedor', fontsize=14)
ax.set_xlabel('Vendedor')
ax.set_ylabel('Ventas ($)')
ax.yaxis.grid(True, linestyle='--', alpha=0.5)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
````
