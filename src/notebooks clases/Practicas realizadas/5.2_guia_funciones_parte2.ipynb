### <span style="color:yellow">**Ejercicio 2** — Funciones que generan listas</span>

Escribí las siguientes funciones que **construyen y retornan** listas:

1. `rango_pares(inicio: int, fin: int) -> list[int]` — retorna lista de números pares entre inicio y fin
2. `fibonacci(n: int) -> list[int]` — retorna los primeros n números de la secuencia de Fibonacci
3. `aplicar_iva(precios: list[float], tasa: float = 0.21) -> list[float]` — retorna nueva lista con IVA aplicado
4. `intercalar(lista_a: list, lista_b: list) -> list` — intercala los elementos de ambas listas
   - `[1, 2, 3]` + `["a", "b", "c"]` → `[1, "a", 2, "b", 3, "c"]`

Probá cada función con ejemplos concretos.

````python
def rango_pares(inicio: int, fin: int) -> list[int]:
   pares = []

   for numero in range(inicio, fin + 1):
       if numero % 2 == 0:
           pares.append(numero)

   return pares

def fibonacci(n: int) -> list[int]:
   """
   Retorna una lista con los primeros n números de la secuencia de Fibonacci.
   """
   if n <= 0:
       return []
   elif n == 1:
       return [0]
      
   # Inicializamos la secuencia con los dos primeros valores estáticos
   secuencia: list[int] = [0, 1]
  
   # Usamos un ciclo para acumular el resto sumando los dos anteriores
   while len(secuencia) < n:
       proximo_numero: int = secuencia[-1] + secuencia[-2]
       secuencia.append(proximo_numero)
      
   return secuencia

def aplicar_iva(precios: list[float], tasa: float = 0.21) -> list[float]:
    precios_finales = []
    for precio in precios:
        precios_finales.append(precio * (1+tasa))
    return precios_finales


def intercalar(lista_a: list, lista_b: list) -> list:
    return [elemento_a for par in zip(lista_a, lista_b) for elemento_a in par]

def intercalar_con_rango(lista_a: list, lista_b: list) -> list:
    resultado = []
   
    for i in range(len(lista_a)):
        resultado.append(lista_a[i])
        resultado.append(lista_b[i])
    return resultado

def intercalar_b(lista_a: list, lista_b: list) -> list:
    nueva_lista = []
    fin_de_lista = []
    if len(lista_a) < len(lista_b):
        lista_mas_corta = len(lista_a) 
        fin_de_lista = lista_b[lista_mas_corta:]
    else: 
        lista_mas_corta = len(lista_b)
        fin_de_lista = lista_a[lista_mas_corta:]
    i = 0
    while i < lista_mas_corta:
        nueva_lista.append(lista_a[i])
        nueva_lista.append(lista_b[i])
        i+=1
    
    return nueva_lista + fin_de_lista

lista_precios = [12, 10, 20, 30]
lista_a = [1, 3, 5, 7]
lista_b = [2, 4, 6, 8, 10, 20]
resultado = intercalar_b(lista_a, lista_b)
print (resultado)
````

4. Explicá en comentarios cuándo conviene cada enfoque y cuál es más seguro de usar en general

> Esta distinción es fundamental en programación: una función pura siempre produce el mismo resultado para los mismos argumentos y no tiene efectos secundarios.

````python
def funcion_impura(a: int):
    a = a + 10

def funcion_pura(a: int):
    return a + 10

numero = 100
funcion_impura(numero)
print(numero)
funcion_pura(numero)
print(numero)

def aplicar_descuento_impuro(lista_precios: list[float], pct: float):
   
    for i in range(len(lista_precios)):
        lista_precios[i] = lista_precios[i] * (1 - pct / 100)
    return 


def aplicar_descuento_puro(precios: list[float], pct: float) -> list[float]:
    return [precio * (1 - pct) for precio in precios]


precios_l = [100.0, 110, 150, 200]
aplicar_descuento_impuro(precios_l, 21)
print(precios_l)
aplicar_descuento_puro(precios_l, 0.21)
print(precios_l)
````

### <span style="color:yellow">**Ejercicio 9** — Integrador: pipeline de procesamiento</span>

Construí un **pipeline** de funciones que procese datos de ventas en etapas. Cada función toma la salida de la anterior:

```python
datos_crudos: list[str] = [
    "Ana,Laptop,2,1500.00",
    "Bruno,Mouse,5,25.00",
    "Ana,Monitor,1,800.00",
    "Carla,Teclado,3,120.00",
    "Bruno,Laptop,1,1500.00",
    "Carla,Mouse,10,25.00",
]
```

1. `parsear(datos_crudos) -> list[dict]` — convierte cada string en dict con clave vendedor/producto/cantidad/precio
2. `calcular_totales(registros) -> list[dict]` — agrega la clave `"total"` a cada dict (cantidad × precio)
3. `resumir_por_vendedor(registros) -> dict[str, float]` — agrupa y suma totales por vendedor
4. `top_vendedor(resumen) -> tuple[str, float]` — retorna el nombre y total del mejor vendedor
5. `ejecutar_pipeline(datos_crudos)` — llama a todas en orden y muestra el resultado final

````python
datos_crudos_orig: list[str] = [
    "Ana,Laptop,2,1500.00",
    "Bruno,Mouse,5,25.00",
    "Ana,Monitor,1,800.00",
    "Carla,Teclado,3,120.00",
    "Bruno,Laptop,1,1500.00",
    "Carla,Mouse,10,25.00",
]

def parsear(datos_crudos: list[str]):
    resultado_total = []
    
    for linea in datos_crudos:
        resultado = {}
        valores = linea.split(",")
        resultado["vendedores"] = valores[0]
        resultado["producto"] = valores[1]
        resultado["cantidad"] = valores[2]
        resultado["precio"] = valores[3]
        resultado_total.append(resultado)
    return resultado_total

def calcular_totales(registros):
    resultado_total = []
    resultado = {}
    for linea in registros:
        resultado = linea
        resultado["total"] = int(resultado["cantidad"]) * float(resultado["precio"])
        resultado_total.append(resultado)
    return resultado_total

def resumir_por_vendedor(registros: list[dict]) -> dict[str, float]:
   resumen = {}

   for registro in registros:
       vendedor = registro["vendedores"]
       total = registro["total"]

       if vendedor not in resumen:
           resumen[vendedor] = 0.0
       resumen[vendedor] += total
   return resumen

def top_vendedor(resumen: dict[str, float]) -> tuple[str, float]:
   """Etapa 4: Identifica al vendedor con la mayor facturación total."""
   mejor_vendedor: str = ""
   max_monto: float = -1.0
   for vendedor, monto in resumen.items():
       if monto > max_monto:
           max_monto = monto
           mejor_vendedor = vendedor
   return mejor_vendedor, max_monto

def ejecutar_pipeline(datos_crudos):
    datos_parseados = parsear(datos_crudos)
    print(datos_parseados)
    datos_con_totales = calcular_totales(datos_parseados)
    print(datos_con_totales)
    resumen_vendedor = resumir_por_vendedor(datos_con_totales)
    print(resumen_vendedor)
    (mejor_vendedor, max_monto) = top_vendedor(resumen_vendedor)
    print("mejor vendedor: ",mejor_vendedor, " con: ", max_monto )

ejecutar_pipeline(datos_crudos_orig)


````
