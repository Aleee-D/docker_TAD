Definí la clase `Empleado` que usa **atributos de clase** para registrar información compartida:

```python
class Empleado:
    empresa: str   = "TechCorp SA"   # atributo de clase
    cantidad: int  = 0               # contador de instancias
```

Atributos de instancia: `nombre`, `puesto`, `salario`

Cada vez que se crea un `Empleado`, el contador `cantidad` debe incrementarse en 1.

Métodos:
1. `__str__` que incluya empresa, nombre y puesto
2. `aumentar_salario(porcentaje: float) -> None`
3. `info_empresa() -> str` — retorna solo el nombre de la empresa (desde el atributo de clase)

Creá 3 empleados y mostrá que `Empleado.cantidad` se actualiza correctamente. Mostrá también la diferencia entre `empleado.empresa` y `Empleado.empresa`.

````python
class Empleado:
    empresa: str  = "TechCorp SA"  # atributo de clase: compartido por todas las instancias
    cantidad: int = 0              # contador de instancias creadas

    def __init__(self, nombre: str, puesto: str, salario: float) -> None:
        self.nombre: str   = nombre
        self.puesto: str   = puesto
        self.salario: float = salario
        Empleado.cantidad = Empleado.cantidad + 1  # modificar el atributo de CLASE

    def cambio_de_puesto(self, nuevo_puesto: str):
        self.puesto = nuevo_puesto

    def __str__(self) -> str:
        return f"{Empleado.empresa} | {self.nombre} — {self.puesto} | ${self.salario:,.0f}"

    def aumentar_salario(self, porcentaje: float) -> None:
        """Aumenta el salario en el porcentaje indicado."""
        self.salario = round(self.salario * (1 + porcentaje / 100), 2)

    def info_empresa(self) -> str:
        """Retorna el nombre de la empresa desde el atributo de clase."""
        return Empleado.empresa

empleado_1 = Empleado("Ana García",   "Desarrolladora", 95000)

print(empleado_1)
empleado_1.cambio_de_puesto("Tester")
print(empleado_1)

````

### **Ejercicio 6** — Relación entre objetos

Modelá una relación entre clases: una `Empresa` tiene una lista de `Empleado`s.

Clase `Empleado`:
- Atributos: `nombre: str`, `puesto: str`, `salario: float`
- Métodos: `__str__`, `aumentar_salario(pct)`

Clase `Empresa`:
- Atributos: `nombre: str`, `empleados: list[Empleado]`
- Métodos:
  1. `contratar(empleado: Empleado) -> None`
  2. `despedir(nombre: str) -> None` — lanza `ValueError` si no existe
  3. `nomina_total() -> float`
  4. `empleado_mejor_pago() -> Empleado`
  5. `dar_aumento_general(pct: float) -> None` — aplica aumento a todos
  6. `__str__` — lista todos los empleados y la nómina total

````python
class Empresa:
   """Clase contenedora que administra una nómina de objetos de tipo Empleado."""


   def __init__(self, nombre: str) -> None:
       """Inicializa la empresa con su nombre corporativo y una nómina vacía."""
       self.nombre: str = nombre
       self.empleados: list[Empleado] = []


   def contratar(self, empleado: Empleado) -> None:
       """Agrega un nuevo objeto Empleado a la nómina corporativa."""
       self.empleados.append(empleado)
       print(f"  [+] {empleado.nombre} fue contratado con éxito.")


   def despedir(self, nombre: str) -> None:
       """Busca y remueve a un empleado por su nombre de forma natural sin break."""
       nombre_buscado: str = nombre.strip().title()
       empleado_a_remover: Empleado | None = None


       for emp in self.empleados:
           if emp.nombre == nombre_buscado:
               empleado_a_remover = emp


       if empleado_a_remover:
           self.empleados.remove(empleado_a_remover)
           print(f"  [-] {nombre_buscado} fue desvinculado de la empresa.")
       else:
           print(f"  ⚠️ Advertencia: No se encontró a '{nombre_buscado}' en la nómina.")


   def nomina_total(self) -> float:
       """Calcula y retorna el costo salarial bruto total de la empresa."""
       total: float = 0.0
       for emp in self.empleados:
           total += emp.salario
       return total


   def empleado_mejor_pago(self) -> Empleado | None:
       """Rastrea y retorna el objeto Empleado con la mayor remuneración."""
       if not self.empleados:
           return None


       # CORRECCIÓN: Asignamos el primer empleado de la lista, no la lista entera
       mejor_pago: Empleado = self.empleados[0]
       for emp in self.empleados:
           if emp.salario > mejor_pago.salario:
               mejor_pago = emp
       return mejor_pago


   def dar_aumento_general(self, pct: float) -> None:
       """Itera sobre la nómina completa aplicando un aumento masivo a cada trabajador."""
       for emp in self.empleados:
           emp.aumentar_salario(pct)


   def __str__(self) -> str:
       """Retorna el reporte de auditoría corporativa de la empresa."""
       if not self.empleados:
           return f"=== EMPRESA: {self.nombre.upper()} ===\n No registra empleados activos."


       lineas: list[str] = [
           "============================================================",
           f" REPORTE DE PERSONAL COMERCIAL — {self.nombre.upper()}",
           "============================================================"
       ]
      
       for emp in self.empleados:
           lineas.append(f"  • {emp}")
          
       lineas.append("------------------------------------------------------------")
       lineas.append(f" Costo mensual de la nómina global: ${self.nomina_total():,.2f}")
      
       lider = self.empleado_mejor_pago()
       if lider:
           lineas.append(f" Máxima remuneración actual:        {lider.nombre} (${lider.salario:,.2f})")
       lineas.append("============================================================")
      
       return "\n".join(lineas)

   def un_empleado(self, numero_empleado):
        return self.empleados[numero_empleado]
# =====================================================================
# PROGRAMA PRINCIPAL (Contiene las llamadas que ejecutan los prints)
# =====================================================================


# 1. Crear la estructura corporativa de la empresa
mi_empresa = Empresa("Antropo-Arg")

emp1 = Empleado("Ana García", "Desarrolladora", 95000.0)
mi_empresa.contratar(emp1)
print(mi_empresa)
print(mi_empresa.un_empleado(0))

````

### <span style="color:yellow">**Ejercicio 9** — Clase `Estadisticas`</span>
Para clase 18/9

Definí la clase `Estadisticas` que almacena una serie de datos numéricos y calcula estadísticas descriptivas.

- Atributo: `datos: list[float]` (se inicializa vacío)

Métodos:
1. `agregar(valor: float) -> None`
2. `agregar_varios(valores: list[float]) -> None`
3. `media() -> float` — lanza `ValueError` si no hay datos
4. `mediana() -> float`
5. `varianza() -> float`
6. `minimo() -> float`
7. `maximo() -> float`
8. `rango() -> float` — diferencia entre máximo y mínimo
9. `resumen() -> None` — imprime todas las estadísticas formateadas
10. `__str__` → `"Serie con 10 datos | Media: 7.50"`

Probá con una serie de ventas y una de temperaturas.

````python
# Tu resolución acá


class Estadisticas:
   """Clase que almacena series numéricas y calcula métricas descriptivas."""


   def __init__(self) -> None:
       """Inicializa el objeto con una lista de datos vacía."""
       self.datos: list[float] = []


   def __str__(self) -> str:
       """Retorna una descripción breve con la cantidad de datos y su media."""
       if not self.datos:
           return "Serie sin datos registrados."
       return f"Serie con {len(self.datos)} datos | Media: {self.media():.2f}"


   def agregar(self, valor: float) -> None:
       """Añade un único valor numérico a la serie de datos."""
       self.datos.append(valor)


   def agregar_varios(self, valores: list[float]) -> None:
       """Añade un lote de valores numéricos a la serie actual."""
       for v in valores:
           self.datos.append(v)


   def media(self) -> float:
       """Calcula el promedio aritmético de la serie. Retorna 0.0 si está vacía."""
       if not self.datos:
           print("  ⚠️ Advertencia: No se pueden calcular métricas en una serie vacía.")
           return 0.0
       return sum(self.datos) / len(self.datos)


   def mediana(self) -> float:
       """Calcula la mediana de la serie de forma pura."""
       if not self.datos:
           return 0.0
      
       # Ordenamos una copia para no alterar las posiciones de la lista original
       datos_ordenados: list[float] = sorted(self.datos)
       n: int = len(datos_ordenados)
       centro: int = n // 2


       if n % 2 != 0:
           return datos_ordenados[centro]
       else:
           return (datos_ordenados[centro - 1] + datos_ordenados[centro]) / 2.0


   def varianza(self) -> float:
       """Calcula la varianza poblacional de la serie."""
       if not self.datos:
           return 0.0
      
       m: float = self.media()
       suma_diferencias_cuadrado: float = 0.0
       for v in self.datos:
           suma_diferencias_cuadrado += (v - m) ** 2
          
       return suma_diferencias_cuadrado / len(self.datos)


   def minimo(self) -> float:
       """Busca el valor mínimo de la serie de forma manual."""
       if not self.datos:
           return 0.0

       min_val: float = self.datos[0]
       for v in self.datos:
           if v < min_val:
               min_val = v
       return min_val


   def maximo(self) -> float:
       """Busca el valor máximo de la serie de forma manual."""
       if not self.datos:
           return 0.0
      
       max_val: float = self.datos[0]
       for v in self.datos:
           if v > max_val:
               max_val = v
       return max_val


   def rango(self) -> float:
       """Calcula la diferencia absoluta entre el valor máximo y el mínimo."""
       if not self.datos:
           return 0.0
       return self.maximo() - self.minimo()


   def resumen(self) -> None:
       """Imprime por pantalla el reporte estadístico completo alineado en columnas."""
       if not self.datos:
           print("No hay datos cargados para generar un resumen.")
           return


       print("=========================================")
       print(f"{'RESUMEN ESTADÍSTICO CONSOLIDADO':^40}")
       print("=========================================")
       print(f"{'Total registros registrados:':<28}{len(self.datos)}")
       print(f"{'Valor mínimo de la serie:':<28}{self.minimo():.2f}")
       print(f"{'Valor máximo de la serie:':<28}{self.maximo():.2f}")
       print(f"{'Rango de dispersión:':<28}{self.rango():.2f}")
       print(f"{'Media aritmética (Promedio):':<28}{self.media():.2f}")
       print(f"{'Mediana de la muestra:':<28}{self.mediana():.2f}")
       print(f"{'Varianza poblacional:':<28}{self.varianza():.2f}")
       print("=========================================\n")




# --- Programa Principal (Simulación con dos series de prueba) ---


# 1. Crear e instanciar la serie de Ventas Comerciales
print(">>> Procesando Serie de Ventas...")
serie_ventas = Estadisticas()
# Cargamos datos de a uno y en lote
serie_ventas.agregar(15000.0)
serie_ventas.agregar_varios([9800.0, 14300.0, 11200.0, 16700.0, 18900.0])


# Imprimir el objeto y su resumen completo
print(serie_ventas)
serie_ventas.resumen()




# 2. Crear e instanciar la serie de Temperaturas Diarias
print(">>> Procesando Serie de Temperaturas...")
serie_temps = Estadisticas()
serie_temps.agregar_varios([22.5, 18.0, 31.2, 15.8, 28.4, 35.1, 12.3, 27.9, 33.6, 19.2])


# Imprimir el objeto y su resumen completo
print(serie_temps)
serie_temps.resumen()

````
