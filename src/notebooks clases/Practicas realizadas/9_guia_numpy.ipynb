1. Creá un array 1D con `np.arange(1, 13)` y reshapealo a:
   - Matriz `(3, 4)`
   - Matriz `(4, 3)`
   - Matriz `(2, 6)`

2. Dadas las matrices de costos y precios de venta para 3 productos en 4 regiones:
```python
costos = np.array([[100, 110, 95, 120],
                   [200, 195, 210, 205],
                   [ 50,  55,  48,  52]])

precios = np.array([[150, 160, 145, 175],
                    [300, 290, 315, 310],
                    [ 80,  85,  75,  82]])
```
   - Calculá la ganancia por producto y región
   - Calculá el margen porcentual: `(precio - costo) / costo * 100`
   - Indicá qué combinaciones tienen margen mayor al 45%

````python
array: no.ndarray = np.arange(1, 13)


array34 = array.reshape(3,4)
array43 = array.reshape(4,3)
array26 = array.reshape(2,6)


print("Matriz (3,4):", array34)


print("Matriz (4,3):", array43)


print("Matriz (2,6):", array26)
costos = np.array([[100, 110, 95, 120],
                   [200, 195, 210, 205],
                   [ 50,  55,  48,  52]])


precios = np.array([[150, 160, 145, 175],
                    [300, 290, 315, 310],
                    [ 80,  85,  75,  82]])


ganancia = costos * precios


print("Ganancia=", ganancia)


margen_pct = (precios - costos) / costos * 100


print("Margen porcentual=", margen_pct.round(2))


mayores= np.sum(margen_pct>45)


print("Margen mayor al 45%: ", mayores)

````
