### <span style="color:yellow">**Ejercicio 2** — `else` y `finally`</span>

Escribí la función `acceder_elemento(lista: list, indice_str: str) -> None` que:

1. Convierte `indice_str` a `int` → puede lanzar `ValueError`
2. Accede a `lista[indice]` → puede lanzar `IndexError`
3. Si **no** hubo error, el bloque `else` imprime el valor encontrado
4. El bloque `finally` **siempre** imprime `"--- fin de operación ---"`

Probá con los tres casos:
```python
datos = [10, 20, 30, 40, 50]
acceder_elemento(datos, "2")    # OK
acceder_elemento(datos, "99")   # IndexError
acceder_elemento(datos, "abc")  # ValueError
```

````python
def acceder_elemento(lista: list, indice_str: str) -> None:
    try:
        indice = int(indice_str)
        valor = lista[indice-1]
    except ValueError:
        print(f" Error {indice_str} no es válido")
    except IndexError:
        print(f" el índice utilizado no es correcto {indice_str}, índice máximo: {len(lista)}")
    else:
        print(f"Se accedió al elemento {valor} con indice {indice}")
    finally:
        print("fin de operación")

datos = [10, 20, 30, 40, 50]
acceder_elemento(datos, "2")    # OK
acceder_elemento(datos, "6")   # IndexError
acceder_elemento(datos, "abc")  # ValueError
````
