### <span style="color:yellow">**Ejercicio 6** — Recorrer lista de diccionarios</span>

Tenés el registro de ventas semanal:

```python
ventas: list[dict] = [
    {"vendedor": "Ana",   "producto": "Laptop",  "monto": 1500.0, "region": "norte"},
    {"vendedor": "Bruno", "producto": "Mouse",   "monto":   25.0, "region": "sur"},
    {"vendedor": "Ana",   "producto": "Monitor", "monto":  800.0, "region": "norte"},
    {"vendedor": "Carla", "producto": "Teclado", "monto":  120.0, "region": "centro"},
    {"vendedor": "Bruno", "producto": "Laptop",  "monto": 1500.0, "region": "sur"},
    {"vendedor": "Carla", "producto": "Monitor", "monto":  800.0, "region": "centro"},
]
```

1. Calculá el total vendido por cada vendedor (dict acumulador)
2. Calculá el total por región
3. Encontrá la venta individual más alta
4. Calculá el ticket promedio
5. Usá un conjunto para listar los productos únicos vendidos

````python
ventas: list[dict] = [
    {"vendedor": "Ana",   "producto": "Laptop",  "monto": 1500.0, "region": "norte"},
    {"vendedor": "Bruno", "producto": "Mouse",   "monto":   25.0, "region": "sur"},
    {"vendedor": "Ana",   "producto": "Monitor", "monto":  800.0, "region": "norte"},
    {"vendedor": "Carla", "producto": "Teclado", "monto":  120.0, "region": "centro"},
    {"vendedor": "Bruno", "producto": "Laptop",  "monto": 1500.0, "region": "sur"},
    {"vendedor": "Carla", "producto": "Monitor", "monto":  800.0, "region": "centro"},
]

por_vendedor: dict[str, float] = {}
por_region:   dict[str, float] = {}
venta_max: float               = 0.0
venta_max_vendedor: str        = ""
total: float                   = 0.0
productos: set[str]            = set()

for v in ventas:
    #Por vendedor
    if v["vendedor"] not in por_vendedor:
        por_vendedor[v["vendedor"]] = v["monto"]
    else:
        por_vendedor[v["vendedor"]] += v["monto"]

    #Por region
    if v["region"] not in por_region:
        por_vendedor[v["region"]] = v["monto"]
    else:
        por_vendedor[v["region"]] += v["monto"]

    productos.add(v["producto"])
    total += v["monto"]

    #Máximos
    if v["monto"] > venta_max:
        venta_max          = v["monto"]
        venta_max_vendedor = v["vendedor"]

promedio = total / len(ventas)

print("Por vendedor: ", por_vendedor)
print("Por región: ", por_region)
print(f"Venta más alta: {venta_max} por {venta_max_vendedor}")
print(f"Ticket promedio: {promedio:.2f}")
print("Productos únicos: ", productos)






````

### <span style="color:yellow">**Ejercicio 10** — Integrador completo: sistema de ranking</span>

Construí un sistema que procese los resultados de un torneo de programación. Cada participante resolvió una serie de problemas con distintos puntajes:

```python
resultados: list[dict] = [
    {"nombre": "Ana",   "puntajes": [100, 85, 90, 70, 95]},
    {"nombre": "Bruno", "puntajes": [60,  75, 80, 55, 70]},
    {"nombre": "Carla", "puntajes": [95, 100, 85, 90, 100]},
    {"nombre": "Diego", "puntajes": [70,  65, 80, 75, 60]},
    {"nombre": "Elena", "puntajes": [80,  90, 85, 95, 88]},
]
```

El sistema debe:

1. Calcular el puntaje total y promedio de cada participante
2. Ordenar el ranking de mayor a menor puntaje total (sin `.sort()`, usá burbuja)
3. Mostrar el podio con medallas (🥇🥈🥉)
4. Calcular el puntaje promedio general del torneo
5. Mostrar qué participantes superaron el promedio general

````python
resultados: list[dict] = [
    {"nombre": "Ana",   "puntajes": [100, 85, 90, 70, 95]},
    {"nombre": "Bruno", "puntajes": [60,  75, 80, 55, 70]},
    {"nombre": "Carla", "puntajes": [95, 100, 85, 90, 100]},
    {"nombre": "Diego", "puntajes": [70,  65, 80, 75, 60]},
    {"nombre": "Elena", "puntajes": [80,  90, 85, 95, 88]},
]

# Calcular totales y promedios
ranking: list[dict] = []
suma_general: float = 0.0

#crear el ranking
for e in resultados:
    total = sum(e["puntajes"])
    promedio = total / len(e["puntajes"])
    suma_general += total
    ranking.append({"nombre": e["nombre"], "total": total, "promedio": promedio})

# Ordenar por total (burbuja)
n = len(ranking)
for i in range(n):
    for j in range(0, n - i - 1):
        if ranking[j]["total"] < ranking[j+1]["total"]:
            ranking[j], ranking[j+1] = ranking[j+1], ranking[j]

prom_general: float = suma_general / n
for r in ranking:
    print(f"  ✓ {r['nombre']} ({r['total']}pts)")
````
