```python
equipo_a: set[str] = {"Ana", "Bruno", "Carla", "Diego", "Elena"}
equipo_b: set[str] = {"Bruno", "Fede", "Gabi", "Diego", "Hugo"}
```

El programa debe:
1. Verificar si hay jugadores en **ambos** equipos (conflicto de inscripción)
2. Si hay conflicto, mostrar los nombres de los jugadores duplicados
3. Verificar si un jugador ingresado por el usuario está en alguno de los dos equipos
4. Mostrar en cuál/cuáles equipos está ese jugador

````python
equipo_a: set[str] = {"Ana", "Bruno", "Carla", "Diego", "Elena"}
equipo_b: set[str] = {"Bruno", "Fede", "Gabi", "Diego", "Hugo"}

#1
grupo_de_repetidos: set[str] = equipo_a & equipo_b

if (len(grupo_de_repetidos) > 0):
    print("Hay jugadores repetidos")
else:
    print("NO hay jugadores repetidos")
#2
print("jugadores duplicados ", grupo_de_repetidos)

#3
usuario = str(input("Ingrese nombre de nuevo jugador"))
if (usuario.title() in equipo_a) or (usuario.title() in equipo_b):
    print("El jugador ya está inscripto")
    if (usuario.title() in equipo_a):
        print("Está en el equipo A")
    if (usuario.title() in equipo_b):
        print("Está en el equipo B")
````
