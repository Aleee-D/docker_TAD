### <span style="color:yellow">**Ejercicio 2** — `while`: adivinar el número</span>

Escribí un juego donde el programa elige un número fijo (no uses `random` todavía) y el usuario tiene que adivinarlo. El programa debe dar pistas:

- Si el número ingresado es mayor: `"Más chico"`
- Si es menor: `"Más grande"`
- Si acierta: `"¡Correcto! Lo lograste en X intentos"`

El ciclo termina **solo** cuando el usuario adivina. Usá una variable de control en el `while`.

````python
parametro: str = input("ingrese un texto")
numero_seleccionado = 1
for letra in parametro:
    numero_seleccionado += (ord(letra) * numero_seleccionado)
numero_seleccionado = numero_seleccionado % 10

#print("El número seleccionado es: ", numero_seleccionado)

respuesta_texto = ""
ingresa_numero: int = -1
intentos = 0
salir = False
while (not salir) and (numero_seleccionado != ingresa_numero):
    respuesta_usuario = input("ingrese un número o 'q' para salir")
    if respuesta_usuario == "q":
        salir = True
    else:
        ingresa_numero = int(respuesta_usuario)
        intentos += 1

        #1
        match ingresa_numero:
            case g if g > numero_seleccionado:
                 respuesta_texto = "Más chico"
            case p if p < numero_seleccionado:
                respuesta_texto = "Más grande"
            case numero_seleccionado:
                respuesta_texto = f"¡Correcto! Lo lograste en {intentos} intentos"
        print(respuesta_texto)

print("fin")
````
