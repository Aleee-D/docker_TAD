### <span style="color:yellow">**Ejercicio 1** — Listas: creación y operaciones básicas</span>

Creá una lista con las ventas mensuales de una sucursal durante el primer semestre:

```
Enero: 125000 | Febrero: 98000 | Marzo: 143000
Abril: 112000 | Mayo: 167000  | Junio: 189000
```

Luego:
1. Mostrá la lista completa
2. Mostrá la venta del primer y del último mes usando indexing
3. Mostrá las ventas del segundo al cuarto mes usando slicing
4. Agregá la venta de Julio: `$201.000` al final
5. Mostrá la cantidad de meses registrados
6. Mostrá la venta máxima, mínima y el total acumulado

````python
ventas: list[tuple[str, int]] = [
    ("Enero",   125000),
    ("Febrero", 98000),
    ("Marzo",   143000),
    ("Abril",   112000),
    ("Mayo",    167000),
    ("Junio",   189000),
]

#1
print("ventas: ", ventas)

#2
venta_primer_mes = ventas[0]
venta_ultimo_mes = ventas[-1]
print(f"Venta del primer mes {venta_primer_mes} y último mes {venta_ultimo_mes}")

#3
ventas_segundo_a_cuarto_mes = ventas[1:3]
print(f"Ventas del segundo y cuarto mes {ventas_segundo_a_cuarto_mes}")

#4
ventas.append(("Julio", 201000))

#5
print(f"Meses registrados {len(ventas)}")

#6
print(f"max {max(ventas, key=lambda x : x[1])} min {min(ventas, key=lambda x : x[1])}")

````

### <span style="color:yellow">**Ejercicio 6** — Diccionarios: lista de diccionarios</span>

Tenés el siguiente catálogo de productos:

```python
catalogo: list[dict] = [
    {"nombre": "Laptop",      "precio": 1500.0, "stock": 10},
    {"nombre": "Mouse",       "precio": 25.0,   "stock": 0},
    {"nombre": "Monitor",     "precio": 800.0,  "stock": 5},
    {"nombre": "Teclado",     "precio": 120.0,  "stock": 30},
    {"nombre": "Auriculares", "precio": 200.0,  "stock": 0},
]
```

1. Mostrá nombre y precio de cada producto
2. Calculá el valor total del inventario (precio × stock)
3. Mostrá los productos sin stock
4. Mostrá el producto más caro
5. Agregá a cada producto la clave `"precio_con_iva"` (precio × 1.21)

````python
catalogo: list[dict] = [
    {"nombre": "Laptop",      "precio": 1500.0, "stock": 10},
    {"nombre": "Mouse",       "precio": 25.0,   "stock": 0},
    {"nombre": "Monitor",     "precio": 800.0,  "stock": 5},
    {"nombre": "Teclado",     "precio": 120.0,  "stock": 30},
    {"nombre": "Auriculares", "precio": 200.0,  "stock": 0},
]
#1
print(f"Nombre: {catalogo[0]["nombre"]:15s} precio: {catalogo[0]["precio"]:.2f}")
print(f"Nombre: {catalogo[1]['nombre']:15s} precio: {catalogo[1]['precio']:.2f}")
print(f"Nombre: {catalogo[2]['nombre']:15s} precio: {catalogo[2]['precio']:.2f}")
print(f"Nombre: {catalogo[3]['nombre']:15s} precio: {catalogo[3]['precio']:.2f}")
print(f"Nombre: {catalogo[4]['nombre']:15s} precio: {catalogo[4]['precio']:.2f}")

#2
total_inventario = 0
total_inventario += catalogo[0]["precio"] * catalogo[0]["stock"]
total_inventario += catalogo[1]['precio'] * catalogo[1]['stock']
total_inventario += catalogo[2]['precio'] * catalogo[2]['stock']
total_inventario += catalogo[3]['precio'] * catalogo[3]['stock']
total_inventario += catalogo[4]['precio'] * catalogo[4]['stock']
print(f"Total inventario: {total_inventario}")

#3
catalogo_sin_stock : list[dict] = []
if(catalogo[0]["stock"]==0) : catalogo_sin_stock.append(catalogo[0])
if(catalogo[1]["stock"]==0) : catalogo_sin_stock.append(catalogo[1])
if(catalogo[2]["stock"]==0) : catalogo_sin_stock.append(catalogo[2])
if(catalogo[3]["stock"]==0) : catalogo_sin_stock.append(catalogo[3])
if(catalogo[4]["stock"]==0) : catalogo_sin_stock.append(catalogo[4])
print(f"Productos sin stock: {catalogo_sin_stock}")

#4
producto_mas_caro : dict = {"precio": 0}
if(producto_mas_caro["precio"]<catalogo[0]["precio"]) : producto_mas_caro = catalogo[0]
if(producto_mas_caro["precio"]<catalogo[1]["precio"]) : producto_mas_caro = catalogo[1]
if(producto_mas_caro["precio"]<catalogo[2]["precio"]) : producto_mas_caro = catalogo[2]
if(producto_mas_caro["precio"]<catalogo[3]["precio"]) : producto_mas_caro = catalogo[3]
if(producto_mas_caro["precio"]<catalogo[4]["precio"]) : producto_mas_caro = catalogo[4]
print(f"Producto más caro: {producto_mas_caro}")

#5
catalogo[0]["precio_mas_iva"] = round(catalogo[0]["precio"] * 1.21, 2)
catalogo[0]["precio_mas_iva"] = round(catalogo[0]["precio"] * 1.105, 2)
#...
print(catalogo)
````
