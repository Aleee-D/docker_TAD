- *Cabecera de archivo (14 bytes):*

    Reservado (4 bytes): Generalmente en ceros (0x00).
    Desplome/Offset (4 bytes): Posición exacta en bytes donde inician los datos de la matriz de píxeles.
- *Cabecera DIB / Información (40 bytes para BITMAPINFOHEADER):*
    Tamaño de cabecera (4 bytes): 40 (0x28 en hexadecimal).
    Ancho y Alto (4+4 bytes): Dimensiones de la imagen en píxeles (números enteros de 32 bits).

- *Matriz de píxeles (Datos ráster):*
   Cada píxel se compone de 4 bytes: el 1.º es Azul (B), el 2.º es Verde (G), el 3.º es Rojo (R) y el 4.º es el canal Alfa (A) de transparencia.
   Por cada pixel existe [r,g,b,a]

````python
class miBMP32bits:
    long_pixel = 4
    cant_total_pixels = 400
    color_oscurecer = 0
    color_rojo = 1
    color_verde = 2
    color_azul = 3
    tamano_encabezado = 54  # Estándar de la estructura de archivos BMP

    def __init__(self, nombre_archivo: str, modo_apertura: str) -> None:
        """Inicializa los buffers y abre el archivo binario manejando excepciones."""
        self.path_to_file: str = nombre_archivo
        self.modo_apertura: str = modo_apertura
        self.posicion_buffer: int = 0
        
        self.datos_encabezado: bytearray = bytearray()
        self.datos_pixeles: bytearray = bytearray()
        self.datos_pixeles_modificados: bytearray = bytearray()
        self.linea: bytes = bytes()

        try:
            # Abrimos el archivo en el modo binario solicitado (rb o wb)
            self.archivo_abierto = open(nombre_archivo, modo_apertura)
        except (FileNotFoundError, PermissionError) as e:
            print(f"❌ Error al abrir el archivo '{nombre_archivo}': {e}")
            self.archivo_abierto = None

    def leer_datos(self) -> None:
        """Lee el encabezado de 54 bytes y el resto de la zona de píxeles en memoria."""
        if self.archivo_abierto and self.modo_apertura == "rb":
            # Nos aseguramos de leer desde el principio del archivo binario
            self.archivo_abierto.seek(0)
            
            # Leemos los primeros 54 bytes del encabezado de la imagen
            self.datos_encabezado = bytearray(self.archivo_abierto.read(self.tamano_encabezado))
            
            # El resto del archivo corresponde a la matriz de píxeles completos
            self.datos_pixeles = bytearray(self.archivo_abierto.read())

    def obtener_prox_linea(self) -> int:
        """Extrae la siguiente línea de píxeles del buffer y avanza el puntero de líneas."""
        # Calculamos cuántos bytes representa una fila completa en la imagen
        bytes_por_linea: int = self.cant_total_pixels * self.long_pixel
        total_lineas: int = len(self.datos_pixeles) // bytes_por_linea
        
        if self.posicion_buffer < total_lineas:
            index_linea: int = self.posicion_buffer * bytes_por_linea
            self.linea = self.datos_pixeles[index_linea : index_linea + bytes_por_linea]
            self.posicion_buffer += 1 
            return self.posicion_buffer - 1
        else:
            return -1

    def numero_linea_actual(self) -> int:
        """Retorna el índice de la línea que se acaba de extraer del buffer."""
        return self.posicion_buffer - 1

    def modificar_linea(self, pos_inicio: int, pos_final: int, color: int) -> bytes:
        """Modifica los componentes de color de los píxeles dentro del rango horizontal dado."""
        # Convertimos la línea a bytearray mutable para poder editarla por índice
        linea_mutable: bytearray = bytearray(self.linea)
        pixel_index: int = pos_inicio * self.long_pixel
        limite_bytes: int = pos_final * self.long_pixel

        while pixel_index <= limite_bytes:
            # Estructura BMP estándar de 32 bits: los bytes vienen en orden B, G, R, A
            match color:
                case self.color_oscurecer:
                    b = max(0, linea_mutable[pixel_index] - 80)
                    g = max(0, linea_mutable[pixel_index + 1] - 80)
                    r = max(0, linea_mutable[pixel_index + 2] - 80)
                case self.color_rojo:
                    b, g, r = 0, 0, 255
                case self.color_verde:
                    b, g, r = 0, 255, 0
                case self.color_azul:
                    b, g, r = 255, 0, 0
                case _:
                    print("error en la selección de color")
                    raise ValueError("error en la selección de color")

            # Asignamos los nuevos valores en orden BGRA
            linea_mutable[pixel_index] = b
            linea_mutable[pixel_index + 1] = g
            linea_mutable[pixel_index + 2] = r
            # linea_mutable[pixel_index + 3] mantiene el canal Alfa intacto
            
            pixel_index += self.long_pixel
            
        self.linea = bytes(linea_mutable)
        return self.linea

    def modificar_pixeles(self, linea_inicial: int, linea_final: int, pos_inicial: int, pos_final: int, color: int) -> None:
        """Orquesta la lectura, procesamiento por líneas y acumulación de píxeles modificados."""
        if self.archivo_abierto and not self.archivo_abierto.closed:
            self.leer_datos()
            self.posicion_buffer = 0
            self.datos_pixeles_modificados = bytearray()
            
            # Ciclo de control natural sin break controlado por el retorno de obtener_prox_linea
            continuar: bool = True
            while continuar:
                indice_linea: int = self.obtener_prox_linea()
                if indice_linea >= 0:
                    # Si el número de línea actual está en el rango vertical, la editamos
                    if linea_inicial <= self.numero_linea_actual() < linea_final:
                        self.modificar_linea(pos_inicial, pos_final, color)
                    # Acumulamos la línea (sea modificada o la original)
                    self.datos_pixeles_modificados.extend(self.linea)
                else:
                    continuar = False
        else:
            print("archivo cerrado o inválido para lectura")

    def encabezado(self) -> bytearray:
        """Retorna el bytearray con los 54 bytes del encabezado original."""
        return self.datos_encabezado

    def pixeles_modificados(self) -> bytearray:
        """Retorna la matriz de píxeles ya procesada."""
        return self.datos_pixeles_modificados

    def copiar_datos_modificados(self, otroMiBMP32bits: 'miBMP32bits') -> None:
        """Copia el encabezado y los píxeles modificados desde otra instancia BMP."""
        self.datos_encabezado = otroMiBMP32bits.encabezado()
        self.datos_pixeles = otroMiBMP32bits.pixeles_modificados()

    def guardar(self) -> None:
        """Escribe físicamente en el disco el encabezado y la zona de píxeles modificados."""
        if self.archivo_abierto and self.modo_apertura == "wb":
            self.archivo_abierto.write(self.datos_encabezado)
            self.archivo_abierto.write(self.datos_pixeles)
            print(f"  ✓ Archivo binario '{self.path_to_file}' guardado con éxito.")
        else:
            print("Error: El archivo no está abierto en modo de escritura ('wb').")

    def __del__(self) -> None:
        """Destructor de clase. Garantiza el cierre del archivo en el sistema operativo."""
        if hasattr(self, 'archivo_abierto') and self.archivo_abierto:
            self.archivo_abierto.close()


# --- Programa Principal / Simulación del Taller de Bigotes ---

path_archivo_mona_lisa = "Mona-Lisa_taller.bmp"
path_archivo_mona_lisa_bigote = "Mona-Lisa_taller_bigote.bmp"

# 1. Abrimos el archivo de origen en modo de lectura binaria ("rb")
archivo_mona_lisa = miBMP32bits(path_archivo_mona_lisa, "rb")

# 2. Abrimos el archivo de destino en modo de escritura binaria ("wb")
archivo_mona_lisa_bigote = miBMP32bits(path_archivo_mona_lisa_bigote, "wb")

# Supongamos que la Gioconda tiene bigotes entre las líneas 180 y 190, y horizontalmente entre píxeles 150 y 250
linea_ini, linea_fin = 180, 190
pos_ini, pos_fin = 150, 250
color_elegido = miBMP32bits.color_oscurecer  # Pintamos un bigote oscuro

print("=== Iniciando Procesamiento Digital de Imagen ===")
# Modificamos los píxeles en la instancia de origen
archivo_mona_lisa.modificar_pixeles(linea_ini, linea_fin, pos_ini, pos_fin, color_elegido)

# Copiamos la estructura a la instancia que se va a guardar
archivo_mona_lisa_bigote.copiar_datos_modificados(archivo_mona_lisa)

# 3. Guardar el archivo definitivo con los bigotes aplicados
archivo_mona_lisa_bigote.guardar()

# Forzamos la destrucción para liberar los descriptores de archivos del sistema
del archivo_mona_lisa
del archivo_mona_lisa_bigote
print("=== Proceso finalizado e hilos de archivos liberados ===")
````
