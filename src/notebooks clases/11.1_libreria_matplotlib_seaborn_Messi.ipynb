## **Matplotlib**

La convención es importar `matplotlib.pyplot` con el alias `plt`. Cada gráfico se construye llamando funciones sobre `plt` y se muestra con `plt.show()`.

**Estructura básica de cualquier gráfico:**
```python
plt.figure(figsize=(ancho, alto))  # tamaño del gráfico (opcional)
plt.tipo_de_grafico(datos)         # definir el gráfico
plt.title("Título")                # título
plt.xlabel("Eje X")               # etiqueta eje X
plt.ylabel("Eje Y")               # etiqueta eje Y
plt.legend()                       # leyenda (si hay varias series)
plt.grid(True)                     # grilla (opcional)
plt.show()                         # mostrar
```

````python
def vectores_goles_Messi(n):
    import csv
    datos_Messi = []
    fechas_goles_i = []
    goles_partido_i = []
    minuto_gol_i = []
    with open("Lionel_Messi_All_Club_Goals.csv", "r", encoding="utf-8") as f:
        reader = csv.reader(f,delimiter=',')
        encabezado: list[str] = next(reader)
        for index, fila in enumerate(reader):
            if (index < n+1):
                #datos_Messi.append(fila)
                fechas_goles_i.append(fila[3]) 
                goles_partido_i.append(int(fila[7].split(":")[0])+int(fila[7].split(":")[1].split(" ")[0]))
                minuto_gol_i.append(eval(fila[9])) 
    
    return [fechas_goles_i, goles_partido_i, minuto_gol_i]

datosMessi = vectores_goles_Messi(100)
print(datosMessi[2])
plt.figure( figsize=(10, 4) )
#plt.plot(meses, ventas, color="red", linestyle='-.', linewidth=0.5, marker='o', markerfacecolor='red',  markeredgecolor='blue')
plt.plot(datosMessi[0], datosMessi[2], color="red", linestyle='-.', linewidth=0.5, marker='o', markerfacecolor='red',  markeredgecolor='blue')

plt.title("Partidos con goles de Messi")
plt.ylabel("goles totales del partido")
plt.xlabel("fecha")

plt.show()
````

#### Barras horizontales y agregar etiquetas de valor

````python
productos: list[str] = ["Laptop", "Mouse", "Monitor", "Teclado", "Auriculares"]
ventas: list[int]    = [4500, 1200, 3200, 800, 1500]

plt.figure(figsize=(9, 5))
barras = plt.barh(productos, ventas, color='steelblue', edgecolor='white')

# Agregar el valor al final de cada barra
for barra in barras:
    ancho: float = barra.get_width()
    plt.text(
        ancho + 50,                     # posición X
        barra.get_y() + barra.get_height() / 2,  # posición Y (centrado)
        str(ancho),
        va='center', fontsize=10
    )

plt.title("Ventas por producto", fontsize=14)
plt.xlabel("Ventas ($)")
plt.grid(axis='x', alpha=0.4)
plt.show()
````

### **Histograma**

Un histograma muestra la **distribución de frecuencias** de un conjunto de datos: divide el rango de valores en intervalos (bins) y cuenta cuántos datos caen en cada uno.

Es útil para entender cómo se distribuyen los datos: si son simétricos, sesgados, si hay valores extremos, etc.

````python
# Generamos datos simulados: notas de 200 alumnos
np.random.seed(42)
notas: np.ndarray = np.random.normal(loc=7.0, scale=1.5, size=200)
notas = np.clip(notas, 1, 10)   # limitar entre 1 y 10

plt.figure(figsize=(9, 5))
plt.hist(notas, bins=15, color='steelblue', edgecolor='white')

plt.title("Distribución de notas", fontsize=14)
plt.xlabel("Nota")
plt.ylabel("Frecuencia")
plt.grid(axis='y', alpha=0.4)
plt.show()

print("Media:  ", round(float(np.mean(notas)), 2))
print("Mediana:", round(float(np.median(notas)), 2))
print("Std:    ", round(float(np.std(notas)), 2))
````

````python
# Comparar dos distribuciones en el mismo histograma
np.random.seed(42)
grupo_a: np.ndarray = np.random.normal(6.5, 1.5, 150)
grupo_b: np.ndarray = np.random.normal(7.5, 1.2, 150)
grupo_a = np.clip(grupo_a, 1, 10)
grupo_b = np.clip(grupo_b, 1, 10)

plt.figure(figsize=(9, 5))
plt.hist(grupo_a, bins=15, alpha=0.6, color='steelblue', label='Grupo A')
plt.hist(grupo_b, bins=15, alpha=0.6, color='tomato',    label='Grupo B')

plt.title("Comparación de distribuciones", fontsize=14)
plt.xlabel("Nota")
plt.ylabel("Frecuencia")
plt.legend()
plt.grid(axis='y', alpha=0.4)
plt.show()
````

### **Subplots: múltiples gráficos en una figura**

`plt.subplots(filas, columnas)` crea una figura con múltiples gráficos organizados en una grilla. Es útil para comparar distintas visualizaciones del mismo conjunto de datos.

````python
np.random.seed(42)
notas: np.ndarray  = np.random.normal(7.0, 1.5, 200)
notas = np.clip(notas, 1, 10)
meses: list[str]   = ["Ene", "Feb", "Mar", "Abr", "May", "Jun"]
ventas: list[int]  = [1500, 1700, 1600, 1900, 1800, 2100]

fig, ejes = plt.subplots(1, 2, figsize=(14, 5))   # 1 fila, 2 columnas

# Gráfico izquierdo: línea de ventas
ejes[0].plot(meses, ventas, marker='o', color='steelblue', linewidth=2)
ejes[0].set_title("Evolución de ventas")
ejes[0].set_xlabel("Mes")
ejes[0].set_ylabel("Ventas ($)")
ejes[0].grid(True, alpha=0.4)

# Gráfico derecho: histograma de notas
ejes[1].hist(notas, bins=15, color='tomato', edgecolor='white')
ejes[1].set_title("Distribución de notas")
ejes[1].set_xlabel("Nota")
ejes[1].set_ylabel("Frecuencia")
ejes[1].grid(axis='y', alpha=0.4)

plt.tight_layout()   # ajusta el espaciado automáticamente
plt.show()
````

````python
# Grilla 2x2
np.random.seed(42)
categorias: list[str] = ["A", "B", "C", "D", "E"]
valores: list[int]    = [3500, 1800, 2200, 1200, 800]
x_linea: np.ndarray   = np.linspace(0, 4 * np.pi, 100)
datos_scatter: tuple  = (np.random.rand(40), np.random.rand(40))

fig, ejes = plt.subplots(2, 2, figsize=(12, 8))

ejes[0, 0].bar(categorias, valores, color='steelblue', edgecolor='white')
ejes[0, 0].set_title("Barras")

ejes[0, 1].plot(x_linea, np.sin(x_linea), color='tomato')
ejes[0, 1].set_title("Línea (seno)")

ejes[1, 0].hist(np.random.normal(0, 1, 500), bins=20, color='seagreen', edgecolor='white')
ejes[1, 0].set_title("Histograma")

ejes[1, 1].scatter(*datos_scatter, color='goldenrod', alpha=0.7)
ejes[1, 1].set_title("Scatter")

for fila in ejes:
    for eje in fila:
        eje.grid(True, alpha=0.3)

plt.suptitle("Panel de gráficos", fontsize=16, y=1.01)
plt.tight_layout()
plt.show()
````

### **`sns.boxplot`: diagrama de caja**

El boxplot muestra en un solo gráfico:

- La **mediana** (línea central)
- El **rango intercuartil** (caja: del percentil 25 al 75)
- Los **bigotes** (hasta 1.5x el rango intercuartil)
- Los **outliers** (puntos fuera de los bigotes)

Es ideal para comparar distribuciones entre grupos.

````python
np.random.seed(42)

# Notas de 3 grupos / cursos
grupo_a: np.ndarray = np.clip(np.random.normal(6.5, 1.5, 80), 1, 10)
grupo_b: np.ndarray = np.clip(np.random.normal(7.5, 1.2, 80), 1, 10)
grupo_c: np.ndarray = np.clip(np.random.normal(5.5, 2.0, 80), 1, 10)

datos: dict = {
    "nota":  list(grupo_a) + list(grupo_b) + list(grupo_c),
    "grupo": ["Grupo A"] * 80 + ["Grupo B"] * 80 + ["Grupo C"] * 80
}

plt.figure(figsize=(8, 5))
sns.boxplot(x=datos["grupo"], y=datos["nota"], palette="muted", hue=datos["grupo"], legend=False)

plt.title("Distribución de notas por grupo", fontsize=14)
plt.xlabel("Grupo")
plt.ylabel("Nota")
plt.show()
````

### **`sns.heatmap`: mapa de calor**

Un heatmap representa una **matriz de valores usando colores**: valores altos se muestran con un color, valores bajos con otro. Es muy usado para visualizar matrices de correlación o tablas de datos numéricos.

````python
# Tabla de ventas: 4 productos x 6 meses
ventas: np.ndarray = np.array([
    [1200, 800,  950,  1500],
    [1350, 750,  1000, 1600],
    [1100, 900,  870,  1400],
    [1500, 1100, 1200, 1800],
    [1400, 980,  1100, 1750],
    [1600, 1050, 1300, 2000],
])

meses:     list[str] = ["Ene", "Feb", "Mar", "Abr", "May", "Jun"]
productos: list[str] = ["Prod A", "Prod B", "Prod C", "Prod D"]

plt.figure(figsize=(8, 5))
sns.heatmap(
    ventas,
    annot=True,           # mostrar el número en cada celda
    fmt='d',              # formato entero
    cmap='YlOrRd',        # paleta de color
    xticklabels=productos,
    yticklabels=meses,
    linewidths=0.5
)

plt.title("Ventas por mes y producto", fontsize=14)
plt.show()
````

### **`sns.barplot`: barras con intervalo de confianza**

Similar al gráfico de barras de Matplotlib, pero agrega automáticamente **barras de error** que muestran la variabilidad de los datos (intervalo de confianza del 95%).

````python
np.random.seed(42)

# Simular puntajes de satisfacción por región (múltiples encuestas por región)
regiones: list[str]  = ["Norte", "Sur", "Centro", "Este", "Oeste"]
puntajes_raw: list   = []
etiquetas_raw: list  = []

medias: list[float] = [7.2, 6.8, 8.1, 7.5, 6.5]

for region, media in zip(regiones, medias):
    n_encuestas: int = np.random.randint(30, 60)
    puntajes_raw.extend(np.clip(np.random.normal(media, 1.0, n_encuestas), 1, 10).tolist())
    etiquetas_raw.extend([region] * n_encuestas)

plt.figure(figsize=(9, 5))
sns.barplot(x=etiquetas_raw, y=puntajes_raw, palette='muted', capsize=0.1)

plt.title("Satisfacción promedio por región", fontsize=14)
plt.xlabel("Región")
plt.ylabel("Puntaje promedio")
plt.ylim(0, 10)
plt.show()
````

### **Resumen**

#### Matplotlib

| Gráfico | Función | Uso típico |
|---|---|---|
| Línea | `plt.plot(x, y)` | Evolución en el tiempo |
| Barras verticales | `plt.bar(x, y)` | Comparar categorías |
| Barras horizontales | `plt.barh(x, y)` | Comparar con etiquetas largas |
| Histograma | `plt.hist(datos, bins=n)` | Distribución de frecuencias |
| Torta | `plt.pie(valores, labels=...)` | Proporciones |
| Scatter | `plt.scatter(x, y)` | Relación entre dos variables |
| Múltiples gráficos | `plt.subplots(f, c)` | Panel comparativo |

#### Personalización Matplotlib

| Elemento | Función |
|---|---|
| Título | `plt.title()` |
| Etiquetas ejes | `plt.xlabel()` / `plt.ylabel()` |
| Leyenda | `plt.legend()` |
| Grilla | `plt.grid(True)` |
| Tamaño | `plt.figure(figsize=(w, h))` |
| Etiquetas eje X | `plt.xticks(posiciones, etiquetas)` |

#### Seaborn

| Gráfico | Función | Ventaja |
|---|---|---|
| Histograma + KDE | `sns.histplot(datos, kde=True)` | Curva de densidad integrada |
| Boxplot | `sns.boxplot(x=grupos, y=valores)` | Muestra mediana, cuartiles, outliers |
| Heatmap | `sns.heatmap(matriz, annot=True)` | Visualizar tablas numéricas |
| Barras con CI | `sns.barplot(x=grupos, y=valores)` | Barra de error automática |
