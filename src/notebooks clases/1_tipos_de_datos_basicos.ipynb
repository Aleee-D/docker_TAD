### **Variables**

Son espacios reservados en la **memoria** principal de una computadora que pueden cambiar de contenido a lo largo de la ejecución de un programa.<br>
Dentro de un programa estos espacios de memoria son referenciados mediante un identificador (nombre de la variable) para poder ser utilizados.

Cada variable es de un tipo de dato determinado, por ejemplo podemos tener variables cuyo contenido es númerico, texto, fechas, etc.

En esta notebook nos centraremos en los tipos de datos más básicos que nos brinda el lenguaje (también llamados tipos de datos simples)


*   Números
*   Cadenas de caracteres
*   Booleanos

En Python, a diferencia de otros lenguajes, no se requiere explicitar el tipo de dato que se guardará en cada variable en el momento de su creación, sino que directamente al asignarle un valor Python determina de qué tipo de dato es.
Esto se conoce como lenguaje debilmente tipado

### **Anotaciones de tipo (Type Hints)**

A partir de **Python 3.5**, el lenguaje incorporó la posibilidad de *anotar* el tipo de dato que se espera en una variable o función. Esto se conoce como **type hints** (pistas de tipo).

**Importante:** Python sigue siendo dinámico. Las anotaciones son **opcionales** — el intérprete no las verifica ni las obliga. Su objetivo es:
- Mejorar la **legibilidad** del código
- Facilitar el trabajo en equipo
- Permitir el uso de herramientas externas de verificación (como `mypy`)

#### Sintaxis básica: anotar variables

La anotación se escribe con `:` después del nombre de la variable, antes del `=`:

````python
# Sin type hints (como vimos antes)
nombre = "Ana"
edad = 25
promedio = 8.75
activo = True

# Con type hints
nombre: str = "Ana"
edad: int = 25
promedio: float = 8.75
activo: bool = True

print(nombre, edad, promedio, activo)
````

#### ¿Python verifica los tipos?

No. Python **no lanza un error** si el tipo asignado no coincide con la anotación:

````python
# Esto NO genera un error en Python (aunque la anotación dice int)
numero: int = "esto es un string"
print(numero)
print(type(numero))  # Veremos que sigue siendo str
````

#### Resumen

| Tipo de dato | Anotación |
|---|---|
| Entero | `int` |
| Decimal | `float` |
| Cadena de caracteres | `str` |
| Booleano | `bool` |
