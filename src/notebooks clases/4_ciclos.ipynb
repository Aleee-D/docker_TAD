En esta notebook vamos a ver los **ciclos** (también llamados bucles o loops): estructuras que permiten repetir un bloque de código múltiples veces.

Python tiene dos tipos de ciclos:

| Ciclo | ¿Cuándo usarlo? |
|---|---|
| `for` | Cuando sabemos de antemano cuántas veces queremos repetir, o cuando recorremos una colección |
| `while` | Cuando repetimos hasta que se cumpla una condición, sin saber cuántas veces |

Además veremos `break`, `continue` y `enumerate()` para controlar y enriquecer los ciclos.

### **Ciclo `for`**

El ciclo `for` recorre una colección elemento por elemento y ejecuta el bloque indentado para cada uno.

```python
for variable in coleccion:
    # bloque que se repite
```

> ⚠️ La variable toma el valor de cada elemento en cada iteración. El nombre que le demos es libre.

#### `for` sobre una lista

````python
valores: list[int] = [0, 1, 2, 3, 4, 5]

for nro in valores:
    print(nro)

print("Esto está afuera del ciclo")
````

````python
# for sobre una lista de strings
palabras: list[str] = ["Arquitectura", "de", "la", "Tecnología", "Informática"]

for palabra in palabras:
    print(palabra)
````

````python
# for sobre una lista con tipos mixtos
lista_mixta: list = [1, 2, 3, "A", "B", "C", [10, 20, 30]]

for elemento in lista_mixta:
    print("Elemento:", elemento, "| Tipo:", type(elemento).__name__)
````

#### `for` sobre un string

Los strings también son iterables: se puede recorrer letra por letra.

````python
palabra: str = "Python"

for letra in palabra:
    print(letra)
````

#### `for` sobre una tupla

````python
coordenadas: tuple[float, float] = (-34.6037, -58.3816)
etiquetas: tuple[str, str]        = ("Latitud", "Longitud")

meses: tuple = ("Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio",
                "Julio", "Agosto", "Septiembre", "Octubre", "Noviembre", "Diciembre")

for mes in meses:
    print(mes)
````

#### `for` sobre un diccionario

Con los métodos `.keys()`, `.values()` e `.items()` podemos recorrer distintas partes del diccionario.

````python
producto: dict[str, str | float | int] = {
    "nombre": "Laptop",
    "precio": 1500.0,
    "stock": 10
}

# Recorrer solo las claves
print("--- Claves ---")
for clave in producto.keys():
    print(clave)

# Recorrer solo los valores
print("--- Valores ---")
for valor in producto.values():
    print(valor)

# Recorrer clave y valor juntos (lo más útil)
print("--- Items ---")
for clave, valor in producto.items():
    print(clave, ":", valor)
````

#### `for` sobre un conjunto

````python
frutas: set[str] = {"manzana", "banana", "naranja", "uva"}

for fruta in frutas:
    print(fruta)

# Nota: el orden no está garantizado en los conjuntos
````

### **`range()`: generar secuencias numéricas**

`range()` genera una secuencia de números. Acepta hasta 3 parámetros:

```python
range(fin)               # desde 0 hasta fin-1
range(inicio, fin)       # desde inicio hasta fin-1
range(inicio, fin, paso) # desde inicio hasta fin-1, de a paso
```

````python
# range con un solo argumento: del 0 al 9
for nro in range(10):
    print(nro, end=" ")  # end=" " imprime en la misma línea
print()  # salto de línea al final
````

````python
# range con inicio y fin
for nro in range(5, 15):
    print(nro, end=" ")
print()
````

````python
# range con paso: de 2 en 2 (números pares)
for nro in range(0, 21, 2):
    print(nro, end=" ")
print()

# range con paso negativo: cuenta regresiva
for nro in range(10, 0, -1):
    print(nro, end=" ")
print()
````

#### Ejemplo aplicado: tabla de multiplicar

````python
numero: int = int(input("Ingresá un número para ver su tabla: "))

for i in range(1, 11):
    resultado: int = numero * i
    print(numero, "x", i, "=", resultado)
````

### **`enumerate()`: iterar con índice y valor**

Cuando necesitamos tanto la **posición** como el **valor** de cada elemento, usamos `enumerate()`. Evita tener que manejar un contador manualmente.

```python
for indice, valor in enumerate(coleccion):
    ...
```

````python
# Sin enumerate (la forma manual, más incómoda)
nombres: list[str] = ["Ana", "Bruno", "Carlos", "Diana"]
indice: int = 0

for nombre in nombres:
    print(indice, nombre)
    indice = indice + 1
````

````python
# Con enumerate (la forma pythónica)
nombres: list[str] = ["Ana", "Bruno", "Carlos", "Diana"]

for indice, nombre in enumerate(nombres):
    print(indice, nombre)
````

````python
# enumerate puede arrancar desde cualquier número
alumnos: list[str] = ["Lucía", "Martín", "Valentina"]

for puesto, alumno in enumerate(alumnos, start=1):
    print("Puesto", puesto, ":", alumno)
````

#### Ejemplo aplicado: mostrar un menú numerado

````python
opciones: list[str] = ["Ver reporte", "Cargar datos", "Exportar", "Salir"]

print("=== MENÚ ===")
for numero, opcion in enumerate(opciones, start=1):
    print(numero, "-", opcion)
````

### **Patrones de acumulación**

Un patrón muy frecuente en Analítica de Negocios es recorrer una colección para **calcular un resultado acumulado**: una suma, un conteo, un máximo, un mínimo.

#### Sumar todos los elementos de una lista

````python
ventas: list[float] = [1500.0, 2300.5, 980.0, 4200.0, 750.25]

total: float = 0.0

for venta in ventas:
    total = total + venta

print("Total de ventas: $", total)
````

#### Contar elementos que cumplen una condición

````python
notas: list[float] = [8.5, 4.0, 7.0, 3.5, 9.0, 6.0, 5.5, 10.0]

aprobados: int   = 0
desaprobados: int = 0

for nota in notas:
    if nota >= 6:
        aprobados = aprobados + 1
    else:
        desaprobados = desaprobados + 1

print("Aprobados:",    aprobados)
print("Desaprobados:", desaprobados)
````

#### Encontrar el máximo y el mínimo

````python
temperaturas: list[float] = [22.5, 18.0, 30.1, 15.3, 27.8, 11.0, 33.4]

maximo: float = temperaturas[0]  # arrancamos con el primer valor
minimo: float = temperaturas[0]

for temp in temperaturas:
    if temp > maximo:
        maximo = temp
    if temp < minimo:
        minimo = temp

print("Temperatura máxima:", maximo)
print("Temperatura mínima:", minimo)
````

#### Ejemplo completo: análisis de ventas por producto

````python
ventas: list[dict] = [
    {"producto": "Laptop",   "monto": 1500.0},
    {"producto": "Mouse",    "monto": 25.0},
    {"producto": "Teclado",  "monto": 80.0},
    {"producto": "Monitor",  "monto": 400.0},
    {"producto": "Auriculares", "monto": 60.0},
]

total: float   = 0.0
cantidad: int  = 0

for venta in ventas:
    print(venta["producto"], "→ $", venta["monto"])
    total    = total + venta["monto"]
    cantidad = cantidad + 1

print("\nTotal vendido:  $", total)
print("Cantidad de productos:", cantidad)
print("Ticket promedio: $", total / cantidad)
````

### **Ciclos anidados**

Un ciclo puede contener otro ciclo en su interior. Por cada iteración del ciclo externo, el ciclo interno se ejecuta completo.

````python
# Tabla de multiplicar del 1 al 5
for i in range(1, 6):
    for j in range(1, 6):
        print(i * j, end="\t")  # \t es tabulación
    print()  # salto de línea al terminar cada fila
````

#### Ejemplo aplicado: recorrer una lista de listas

````python
# Calificaciones de 3 alumnos, cada uno con 4 notas
calificaciones: list[list[float]] = [
    [8.0, 7.5, 9.0, 6.5],
    [5.0, 6.0, 4.5, 7.0],
    [9.5, 10.0, 8.5, 9.0],
]
nombres: list[str] = ["Ana", "Bruno", "Carla"]

for indice, notas in enumerate(calificaciones):
    total: float   = 0.0
    for nota in notas:
        total = total + nota
    promedio: float = total / len(notas)
    estado: str     = "Aprobado" if promedio >= 6 else "Desaprobado"
    print(nombres[indice], "→ promedio:", promedio, "|", estado)
````

### **Ciclo `while`**

El ciclo `while` repite un bloque de código **mientras una condición sea verdadera**. A diferencia del `for`, no recorre una colección sino que sigue hasta que la condición falla.

```python
while condicion:
    # bloque que se repite
```

> ⚠️ Si la condición nunca se vuelve `False`, el ciclo es infinito y el programa se cuelga. Siempre hay que asegurarse de que algo dentro del ciclo modifique la condición.

````python
# Contador clásico con while
a: int = 0

while a < 10:
    print("valor actual de a:", a)
    a = a + 1  # ← esto hace que eventualmente la condición sea False

print("Por acá sigue después del ciclo")
````

#### Ejemplo aplicado: búsqueda en una lista

````python
nombres: list[str] = ["Pedro", "Juan", "Pablo", "Pepe", "Rodrigo", "Mariano", "Santiago"]
nombre_a_buscar: str = input("¿Qué nombre querés buscar? ")

encontrado: bool = False
indice: int      = 0

while not encontrado and indice < len(nombres):
    if nombres[indice] == nombre_a_buscar:
        encontrado = True
    else:
        indice = indice + 1

if encontrado:
    print("Encontramos", nombre_a_buscar, "en la posición", indice)
else:
    print(nombre_a_buscar, "no está en la lista")
````

### **`break` y `continue`: controlar el flujo del ciclo**

| Sentencia | Efecto |
|---|---|
| `break` | **Interrumpe** el ciclo inmediatamente y sale de él |
| `continue` | **Saltea** la iteración actual y pasa a la siguiente |

# Aunque se los expliquemos para que sepan que existen, no se puede utilizar break en los exámenes

#### `break`: salir del ciclo antes de que termine

````python
# Buscar el primer número negativo en una lista
numeros: list[int] = [5, 8, 3, -2, 7, -1, 4]

for nro in numeros:
    if nro < 0:
        print("Primer negativo encontrado:", nro)
        break  # salimos del ciclo ni bien lo encontramos
    print("Revisando:", nro)

print("Fin del ciclo")
````

````python
# break también funciona con while
intentos: int       = 0
max_intentos: int   = 3
password_real: str  = "1234"

while intentos < max_intentos:
    password: str = input("Ingresá tu contraseña: ")
    intentos = intentos + 1

    if password == password_real:
        print("✓ Acceso concedido")
        break
    else:
        restantes: int = max_intentos - intentos
        if restantes > 0:
            print("✗ Contraseña incorrecta. Intentos restantes:", restantes)
        else:
            print("✗ Cuenta bloqueada")
````

#### `continue`: saltear una iteración

````python
# Imprimir solo los números impares
for nro in range(1, 11):
    if nro % 2 == 0:   # si es par...
        continue       # ...saltear esta iteración
    print(nro)
````

````python
# Procesar una lista ignorando valores inválidos
datos: list = [10, -5, 30, 0, 15, -3, 50]

total: float = 0.0
procesados: int = 0

for dato in datos:
    if dato <= 0:
        print("Valor ignorado:", dato)
        continue       # saltear valores no positivos
    total = total + dato
    procesados = procesados + 1

print("Suma de valores válidos:", total)
print("Cantidad procesada:", procesados)
````

### **Patrón `while True` con `break`**

`while True` crea un ciclo que en principio nunca termina. Lo usamos cuando queremos que el ciclo siga hasta que el usuario decida salir, usando `break` para terminar.

Es el patrón clásico para **menús interactivos** y **validación de entradas**.

#### Validación de entrada: asegurarse de que el usuario ingrese un número válido

````python
while True:
    entrada: str = input("Ingresá un número entre 1 y 10: ")

    # Verificamos que sea un dígito
    if not entrada.isdigit():
        print("Error: ingresá solo números")
        continue

    numero: int = int(entrada)

    if numero < 1 or numero > 10:
        print("Error: el número debe estar entre 1 y 10")
        continue

    # Si llegamos acá, el dato es válido
    print("Número válido:", numero)
    break
````

#### Menú interactivo

````python
while True:
    print("\n=== MENÚ ===")
    print("1 - Ver saldo")
    print("2 - Depositar")
    print("3 - Retirar")
    print("4 - Salir")

    opcion: str = input("Elegí una opción: ")

    if opcion == "1":
        print("Tu saldo es: $1000")
    elif opcion == "2":
        print("Depósito registrado")
    elif opcion == "3":
        print("Retiro registrado")
    elif opcion == "4":
        print("Hasta luego!")
        break
    else:
        print("Opción no válida, intentá de nuevo")
````

### **Resumen**

| Herramienta | Uso principal |
|---|---|
| `for` | Recorrer listas, tuplas, dicts, sets, strings |
| `range()` | Generar secuencias numéricas con inicio, fin y paso |
| `enumerate()` | Recorrer con índice y valor al mismo tiempo |
| `while` | Repetir hasta que una condición sea False |
| `while True` + `break` | Menús, validación de entrada |
| `break` | Salir del ciclo antes de que termine |
| `continue` | Saltear una iteración y seguir con la próxima |
| Ciclos anidados | Recorrer estructuras de múltiples niveles |

````python
# Mini repaso de sintaxis

# for sobre lista
for x in [1, 2, 3]:
    print(x)

# for con range
for i in range(0, 6, 2):
    print(i, end=" ")
print()

# for con enumerate
for i, v in enumerate(["a", "b", "c"], start=1):
    print(i, v)

# while
n: int = 3
while n > 0:
    print(n)
    n = n - 1
````
