#<center>Tipo de Datos complejos</center>

En esta notebook vamos a ver los **tipos de datos complejos** de Python: estructuras que permiten agrupar múltiples valores en una sola variable.

| Estructura | Símbolo | Ordenada | Mutable | Claves | Duplicados |
|---|---|---|---|---|---|
| Lista | `[ ]` | ✓ | ✓ | ✗ | ✓ |
| Tupla | `( )` | ✓ | ✗ | ✗ | ✓ |
| Diccionario | `{ clave: valor }` | ✓* | ✓ | ✓ | ✗ (claves) |
| Conjunto | `{ }` | ✗ | ✓ | ✗ | ✗ |

> *Los diccionarios mantienen el orden de inserción desde Python 3.7

---
### **Listas**

listas_1.jpg

Una lista es una colección **ordenada y mutable** de elementos, encerrada entre corchetes `[ ]`.

- Los elementos se separan por comas
- Pueden contener distintos tipos de datos
- Se pueden agregar, eliminar y modificar elementos
- El primer elemento está en el índice `0`, el último en el índice `-1`

````python
# Crear listas
lista_numeros: list = [1, 2, 3, 4]
lista_super: list  = ["Yerba", "Café", "Galletitas"]
lista_mixta: list  = [1, "A", 3.1416, True, False]

print(lista_numeros)
print(lista_super)
print(lista_mixta)
````

````python
# Una lista puede contener otras listas (sublistas)
lista_con_sublista: list = [1, "A", 3.1416, [10, 20, 30]]
print(lista_con_sublista)
print("Sublista:", lista_con_sublista[3])       # acceder a la sublista
print("Elemento de sublista:", lista_con_sublista[3][1])  # elemento dentro de la sublista
````

#### Acceder a elementos

````python
lista: list = [10, 20, 30, 40, 50]

print("Primer elemento:",   lista[0])   # 10
print("Tercer elemento:",   lista[2])   # 30
print("Último elemento:",   lista[-1])  # 50
print("Penúltimo elemento:", lista[-2]) # 40
````

#### Slicing: obtener un rango de elementos

````python
lista: list = [10, 20, 30, 40, 50]

print(lista[1:3])   # [20, 30]  → desde índice 1 hasta antes del 3
print(lista[:3])    # [10, 20, 30]  → desde el inicio hasta antes del 3
print(lista[2:])    # [30, 40, 50]  → desde índice 2 hasta el final
print(lista[:])     # [10, 20, 30, 40, 50]  → copia completa
````

#### Largo de una lista

````python
lista: list = [10, 20, 30, 40, 50]
print("Cantidad de elementos:", len(lista))  # 5
````

#### Modificar elementos

````python
lista: list = [1, 2, 3, 4]
lista[1] = 99
print(lista)  # [1, 99, 3, 4]
````

#### Agregar elementos

````python
lista: list = [1, 2, 3, 4]

# .append() agrega un elemento al final
lista.append(5)
print("Después de append:", lista)

# También se pueden concatenar listas con +
lista2: list = lista + [6, 7, 8]
print("Concatenada:", lista2)
````

#### Eliminar elementos

````python
lista_a: list = [1, 2, 3, 4, 3]

# .remove() elimina la PRIMERA ocurrencia del valor indicado
lista_a.remove(3)
print("Después de remove(3):", lista_a)  # [1, 2, 4, 3]

# .pop() elimina el elemento en la POSICIÓN indicada
lista_a.pop(0)
print("Después de pop(0):", lista_a)      # [2, 4, 3]
````

#### Ordenar listas

````python
lista: list = [5, 2, 8, 1, 9, 3]

lista.sort()                     # ordena de menor a mayor (modifica la lista original)
print("Ascendente:", lista)

lista.sort(reverse=True)         # ordena de mayor a menor
print("Descendente:", lista)
````

#### Unir listas

````python
lista_a: list = [1, 2, 3, 4]
lista_b: list = [5, 6, 7, 8]

lista_c: list = lista_a + lista_b
print(lista_c)
````

#### Verificar si un elemento existe: operador `in`

El operador `in` devuelve `True` si el elemento se encuentra en la lista, y `False` si no. Funciona con todas las estructuras que veremos en esta notebook.

````python
lista_super: list = ["Yerba", "Café", "Galletitas"]

print("Yerba" in lista_super)       # True
print("Azúcar" in lista_super)      # False
print("Azúcar" not in lista_super)  # True
````

#### Ejemplo aplicado: verificar si un producto existe antes de agregarlo

````python
carrito: list = ["Yerba", "Café", "Galletitas"]
producto: str = input("¿Qué producto querés agregar? ")

if producto in carrito:
    print(producto, "ya está en el carrito")
else:
    carrito.append(producto)
    print("Producto agregado. Carrito:", carrito)
````

#### Error frecuente: acceder a un índice que no existe

````python
lista: list = [1, 2, 3, 4]

# Esto genera un IndexError:
# print(lista[10])
````

### **Tuplas**

Las tuplas son como las listas, pero **inmutables**: una vez creadas, no se pueden modificar.

Se definen con paréntesis `( )` y son útiles para representar datos que no deben cambiar, como coordenadas, fechas, o valores de configuración.

````python
# Crear tuplas
tupla_numeros: tuple = (1, 2, 3, 4, 5)
tupla_mixta: tuple   = (1, "A", 3.1416, True, False)
coordenadas: tuple   = (-34.6037, -58.3816)  # latitud, longitud de Buenos Aires

print(tupla_numeros)
print(tupla_mixta)
print("Latitud:", coordenadas[0], "| Longitud:", coordenadas[1])
print(type(tupla_numeros))
````

#### Acceder a elementos (igual que en listas)

````python
tupla: tuple = (10, 20, 30, 40, 50)

print(tupla[0])    # 10 - primer elemento
print(tupla[-1])   # 50 - último elemento
print(tupla[1:3])  # (20, 30) - slicing
print(len(tupla))  # 5 - largo
````

#### Las tuplas son inmutables: no se pueden modificar

````python
tupla: tuple = (1, 2, 3, 4)

# Esto genera un TypeError:
# tupla[0] = 99
````

#### Verificar si un elemento existe

````python
meses_del_año: tuple = ("Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio",
                         "Julio", "Agosto", "Septiembre", "Octubre", "Noviembre", "Diciembre")

print("Marzo" in meses_del_año)     # True
print("Treceavo" in meses_del_año)  # False
````

#### Convertir entre lista y tupla

````python
lista: list   = [1, 2, 3, 4]
tupla: tuple  = (10, 20, 30)

lista_a_tupla: tuple = tuple(lista)   # convertir lista → tupla
tupla_a_lista: list  = list(tupla)    # convertir tupla → lista

print(lista_a_tupla, type(lista_a_tupla))
print(tupla_a_lista, type(tupla_a_lista))
````

#### ¿Cuándo usar lista y cuándo tupla?

- Usá una **lista** cuando necesitás agregar, quitar o modificar elementos (ej: un carrito de compras)
- Usá una **tupla** cuando los datos no deben cambiar (ej: coordenadas, días de la semana, configuraciones fijas)

### **Diccionarios**

data_type_diccionarios_1.PNG

Un diccionario almacena pares **clave → valor**, encerrados entre llaves `{ }`.

- Se accede a cada valor usando su clave (no por posición numérica)
- Las claves deben ser únicas
- Son mutables: se pueden agregar, modificar y eliminar entradas

Son ideales para representar entidades del mundo real: un producto, un cliente, un empleado.

````python
# Crear un diccionario
cliente: dict = {
    "nombre": "Santiago",
    "edad": 25,
    "ciudad": "Buenos Aires"
}

print(cliente)
````

#### Acceder a valores

````python
cliente: dict = {"nombre": "Santiago", "edad": 25, "ciudad": "Buenos Aires"}

print(cliente["nombre"])   # Santiago
print(cliente["edad"])     # 25
````

#### Error frecuente: acceder a una clave que no existe

````python
cliente: dict = {"nombre": "Santiago", "edad": 25}

# Esto genera un KeyError:
# print(cliente["telefono"])
````

#### Modificar y agregar elementos

````python
cliente: dict = {"nombre": "Mariano", "edad": 42}

# Modificar un valor existente
cliente["nombre"] = "Pepe"

# Agregar nuevas claves
cliente["ciudad"]    = "CABA"
cliente["direccion"] = "Av. Montes de Oca 1234"

print(cliente)
````

#### Eliminar elementos

````python
cliente: dict = {"nombre": "Karina", "edad": 25, "ciudad": "Rosario"}
print("Antes:", cliente)

cliente.pop("edad")
print("Después de pop('edad'):", cliente)
````

#### Métodos esenciales: `.keys()`, `.values()`, `.items()`

Estos tres métodos son fundamentales para trabajar con diccionarios:

````python
producto: dict = {
    "nombre": "Laptop",
    "precio": 1500.0,
    "stock": 10
}

# .keys()   → devuelve todas las claves
print("Claves:", list(producto.keys()))

# .values() → devuelve todos los valores
print("Valores:", list(producto.values()))

# .items()  → devuelve pares (clave, valor) como tuplas
print("Items:", list(producto.items()))
````

#### Verificar si una clave existe: operador `in`

````python
cliente: dict = {"nombre": "Santiago", "edad": 25}

print("nombre" in cliente)     # True
print("telefono" in cliente)   # False

# Uso práctico: agregar una clave solo si no existe
if "telefono" not in cliente:
    cliente["telefono"] = "Sin datos"

print(cliente)
````

#### Diccionarios anidados y con listas

````python
# Un diccionario puede contener otros diccionarios o listas como valores
empresa: dict = {
    "nombre": "TechCorp",
    "empleados": 150,
    "sede_principal": {
        "ciudad": "Buenos Aires",
        "pais": "Argentina"
    },
    "productos": ["Software", "Consultoría", "Soporte"]
}

print(empresa["nombre"])
print(empresa["sede_principal"]["ciudad"])   # acceder a diccionario anidado
print(empresa["productos"][0])               # acceder a lista dentro del diccionario
````

#### Ejemplo aplicado: ficha de un alumno

````python
alumno: dict = {
    "nombre": "Lucía",
    "legajo": 12345,
    "nota_final": 8.5
}

if "nota_final" in alumno:
    estado: str = "Aprobado" if alumno["nota_final"] >= 6 else "Desaprobado"
    print(alumno["nombre"], "-", estado, "con", alumno["nota_final"])
else:
    print("El alumno aún no tiene nota final")
````

### **Conjuntos (Sets)**

conjuntos_reduc.png

Un conjunto es una colección **sin orden y sin duplicados**, encerrada entre llaves `{ }`.

- No permite elementos repetidos (los elimina automáticamente)
- No tiene índices (no se puede acceder por posición)
- Son muy útiles para operaciones matemáticas de conjuntos: unión, intersección, diferencia

````python
# Los duplicados se eliminan automáticamente
lista_con_repetidos: list = [1, 1, 2, 2, 3, 3, 3, 4]
conjunto: set             = {1, 1, 2, 2, 3, 3, 3, 4}

print("Lista:",    lista_con_repetidos)   # conserva duplicados
print("Conjunto:", conjunto)              # elimina duplicados
print(type(conjunto))
````

#### Agregar y eliminar elementos

````python
frutas: set = {"manzana", "banana", "naranja"}

frutas.add("uva")          # agregar elemento
print("Con uva:", frutas)

frutas.discard("banana")   # eliminar elemento (no da error si no existe)
print("Sin banana:", frutas)
````

#### Verificar si un elemento existe

````python
frutas: set = {"manzana", "banana", "naranja"}

print("manzana" in frutas)   # True
print("kiwi" in frutas)      # False
````

#### Operaciones matemáticas con conjuntos

Esta es la principal ventaja de los conjuntos: permiten hacer operaciones de la **teoría de conjuntos** de forma muy simple.

````python
grupo_a: set = {"Ana", "Bruno", "Carlos", "Diana"}
grupo_b: set = {"Carlos", "Diana", "Elena", "Fabio"}

# Unión (|): todos los elementos de ambos conjuntos, sin repetir
print("Unión:", grupo_a | grupo_b)

# Intersección (&): solo los elementos que están en AMBOS
print("Intersección:", grupo_a & grupo_b)

# Diferencia (-): elementos que están en A pero NO en B
print("Solo en grupo A:", grupo_a - grupo_b)
print("Solo en grupo B:", grupo_b - grupo_a)

# Diferencia simétrica (^): elementos que están en uno u otro, pero no en ambos
print("En uno solo:", grupo_a ^ grupo_b)
````

#### Caso de uso clásico: eliminar duplicados de una lista

````python
# Una técnica muy común en Analítica de Negocios: convertir a set para eliminar duplicados
ventas_diarias: list = ["Laptop", "Mouse", "Laptop", "Teclado", "Mouse", "Mouse"]
print("Con duplicados:", ventas_diarias)

productos_unicos: list = list(set(ventas_diarias))
print("Sin duplicados:", productos_unicos)
````

#### Ejemplo aplicado: clientes en común entre dos sucursales

````python
clientes_sucursal_norte: set = {"García", "López", "Martínez", "Rodríguez"}
clientes_sucursal_sur: set   = {"López", "Fernández", "Rodríguez", "Sánchez"}

en_ambas:     set = clientes_sucursal_norte & clientes_sucursal_sur
solo_en_norte: set = clientes_sucursal_norte - clientes_sucursal_sur
todos:        set = clientes_sucursal_norte | clientes_sucursal_sur

print("Clientes en ambas sucursales:", en_ambas)
print("Clientes solo en sucursal norte:", solo_en_norte)
print("Total de clientes únicos:", len(todos))
````

### **Anotaciones de tipo para estructuras complejas (Type Hints)**

En la notebook anterior vimos cómo anotar variables simples (`int`, `str`, `float`, `bool`). Podemos hacer lo mismo con las estructuras complejas.

La sintaxis básica es simplemente usar el nombre del tipo:

````python
# Anotaciones básicas
mi_lista:  list  = [1, 2, 3]
mi_tupla:  tuple = (10, 20, 30)
mi_dicc:   dict  = {"nombre": "Ana", "edad": 21}
mi_set:    set   = {"rojo", "verde", "azul"}

print(type(mi_lista))
print(type(mi_tupla))
print(type(mi_dicc))
print(type(mi_set))
````

#### Ser más específico: indicar el tipo de los elementos internos

Desde **Python 3.9** podemos indicar también el tipo de los elementos que contiene la estructura, usando corchetes `[ ]` después del tipo:

```python
list[int]        # lista de enteros
tuple[str, ...]  # tupla de strings
dict[str, float] # diccionario con claves str y valores float
set[str]         # conjunto de strings
```

````python
# Listas con tipo de elemento especificado
notas: list[float]       = [8.5, 7.0, 9.5, 6.0]
nombres: list[str]       = ["Ana", "Bruno", "Carlos"]
flags: list[bool]        = [True, False, True]

print(notas)
print(nombres)
````

````python
# Diccionario con tipos de clave y valor especificados
alumno: dict[str, str]          = {"nombre": "Lucía", "ciudad": "CABA"}
precios: dict[str, float]       = {"Laptop": 1500.0, "Mouse": 25.0}
inventario: dict[str, int]      = {"Yerba": 50, "Café": 30}

print(precios)
print(inventario)
````

````python
# Conjunto con tipo especificado
clientes_vip: set[str]    = {"García", "López", "Martínez"}
numeros_pares: set[int]   = {2, 4, 6, 8, 10}

print(clientes_vip)
````

#### Recordatorio importante

Como vimos antes, Python **no verifica** estas anotaciones en tiempo de ejecución. Son una ayuda visual para el programador, no una restricción del lenguaje.

```python
# Esto NO genera error aunque la anotación dice list[int]
numeros: list[int] = [1, "dos", 3.0, True]
print(numeros)  # Python lo acepta igual
```

````python
# Ejemplo completo: ficha de un producto con type hints detallados
producto: dict[str, str | float | int] = {
    "nombre": "Laptop",
    "precio": 1500.0,
    "stock": 10,
    "categoria": "Tecnología"
}

etiquetas: set[str]      = {"oferta", "nuevo", "destacado"}
historial: list[float]   = [1200.0, 1350.0, 1500.0]  # evolución del precio

print("Producto:", producto["nombre"])
print("Precio actual:", producto["precio"])
print("Etiquetas:", etiquetas)
print("Historial de precios:", historial)
````

| Necesito... | Usar |
|---|---|
| Una colección que puedo modificar | `list` |
| Una colección que NO debe cambiar | `tuple` |
| Asociar etiquetas/nombres a valores | `dict` |
| Eliminar duplicados | `set` |
| Comparar grupos de elementos | `set` |
| Representar una entidad (alumno, producto...) | `dict` |
| Guardar una secuencia de resultados | `list` |

````python
# Resumen de sintaxis
mi_lista:  list  = [1, 2, 3]                            # mutable, ordenada
mi_tupla:  tuple = (1, 2, 3)                            # inmutable, ordenada
mi_dicc:   dict  = {"clave": "valor"}                   # clave-valor, mutable
mi_set:    set   = {1, 2, 3}                            # sin duplicados, sin orden

print(type(mi_lista))
print(type(mi_tupla))
print(type(mi_dicc))
print(type(mi_set))
````
