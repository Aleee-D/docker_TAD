#<center>Sentencias condicionales</center>

En esta notebook vamos a ver las **sentencias condicionales** de Python: estructuras que permiten que el programa tome decisiones y ejecute distintas acciones según se cumplan o no ciertas condiciones.

Veremos:
- Tipos de datos booleanos
- Operadores relacionales y lógicos
- `if` / `else` / `elif`
- Condicionales anidados
- Expresión ternaria (inline if)
- `match / case` (Python 3.10+)

---
### **Tipos de datos booleanos**

booleans_1.PNG

Los booleanos solo pueden tomar dos valores: `True` (verdadero) o `False` (falso).
Son fundamentales para las sentencias condicionales, ya que toda condición se evalúa como uno de estos dos valores.

> **Nota:** en Python, `True` equivale numéricamente a `1` y `False` a `0`.

````python
aprobado = True
tiene_deuda = False

print(aprobado)
print(tiene_deuda)

# Internamente True=1 y False=0
print(0 + aprobado)       # 1
print(0 + tiene_deuda)    # 0
````

---
### **Operadores relacionales**

Permiten comparar dos valores. El resultado siempre es un booleano (`True` o `False`).

| Operador | Descripción | Ejemplo |
|---|---|---|
| `==` | Igual a | `5 == 5` → `True` |
| `!=` | Distinto de | `5 != 3` → `True` |
| `>` | Mayor que | `7 > 3` → `True` |
| `<` | Menor que | `2 < 9` → `True` |
| `>=` | Mayor o igual | `5 >= 5` → `True` |
| `<=` | Menor o igual | `4 <= 6` → `True` |

````python
nro1 = 1
nro2 = 2

print(nro1 > nro2)    # False: 1 no es mayor que 2
print(nro2 > nro1)    # True:  2 sí es mayor que 1
print(nro2 >= nro1)   # True:  2 es mayor o igual que 1
print(nro2 == nro1)   # False: 2 no es igual a 1
print(nro2 != nro1)   # True:  2 sí es distinto de 1
````

````python
# También podemos guardar el resultado en una variable
condicion1 = nro2 >= nro1
print(condicion1)          # True
print(type(condicion1))    # <class 'bool'>
````

⚠️ **Cuidado con las comparaciones de texto:** Python distingue mayúsculas de minúsculas.

````python
palabra1 = "casa"
palabra2 = "CASA"
palabra3 = "casa"

print(palabra1 == palabra2)   # False: diferente capitalización
print(palabra1 == palabra3)   # True:  son idénticas
````

````python
print("palabra1 a minúscula:", palabra1.lower() )
print("palabra2 a minúscula:", palabra2.lower() )

palabra1.lower() == palabra2.lower()
````

````python
palabra1_minuscula=palabra1.lower()
palabra2_minuscula=palabra2.lower()

palabra1_minuscula==palabra2_minuscula
````

### **Operadores lógicos**

Permiten combinar o modificar condiciones booleanas.

| Operador | Descripción |
|---|---|
| `and` | True **solo si ambas** condiciones son True |
| `or` | True si **al menos una** condición es True |
| `not` | **Invierte** el valor booleano |

````python
# AND: ambas deben ser True
print("--- AND ---")
print(True  and True)    # True
print(True  and False)   # False
print(False and True)    # False
print(False and False)   # False
````

````python
# OR: al menos una debe ser True
print("--- OR ---")
print(True  or True)    # True
print(True  or False)   # True
print(False or True)    # True
print(False or False)   # False
````

````python
# NOT: invierte el valor
print("--- NOT ---")
print(not True)    # False
print(not False)   # True
````

````python
# Ejemplo combinado: un alumno aprueba si tiene nota >= 6 Y no tiene deuda
nota: float = 7.5
tiene_deuda: bool = False

puede_aprobar = nota >= 6 and not tiene_deuda
print(puede_aprobar)   # True
````

### **Sentencias condicionales: `if` / `else`**

La estructura `if` permite ejecutar un bloque de código **solo si** se cumple una condición.

```
if condicion:
    # código si la condición es True
else:
    # código si la condición es False  (opcional)
```

> ⚠️ **Indentación:** Python usa la indentación (espacios o tab) para delimitar los bloques. El código dentro de un `if` debe estar indentado.

````python
# Ejemplo mínimo: if sin else
a: int = 2
b: int = 2

if a == b:
    print("son iguales")

print("esto se ejecuta siempre")
````

````python
# if + else
a: int = 1
b: int = 2

if a == b:
    print("son iguales")
else:
    print("son distintos")

print("Sigue por acá")
````

#### Ejemplo aplicado: verificar si un número es par o impar

````python
numero: int = 7

if numero % 2 == 0:
    print(numero, "es par")
else:
    print(numero, "es impar")
````

#### Ejemplo aplicado: aprobación de un alumno

````python
nota: float = float(input("Ingresá tu nota (0 a 10): "))

if nota >= 6:
    print("Aprobado ✓")
else:
    print("Desaprobado ✗")
````

---
### **`elif`: múltiples condiciones**

Cuando hay más de dos casos posibles, usamos `elif` (abreviación de *else if*). Python evalúa las condiciones **en orden** y ejecuta el primer bloque cuya condición sea `True`.

```
if condicion_1:
    ...
elif condicion_2:
    ...
elif condicion_3:
    ...
else:
    ...  # si ninguna condición anterior fue True
```

````python
# Detectar si un número es positivo, negativo o cero
a: int = -1

if a > 0:
    print("a es positivo")
elif a < 0:
    print("a es negativo")
else:
    print("a es cero")

print("Esto se ejecuta siempre")
````

````python
# Clasificar una nota en rangos
nota: float = 8.5

if nota < 4:
    calificacion: str = "Insuficiente"
elif nota < 6:
    calificacion = "Regular"
elif nota < 8:
    calificacion = "Bueno"
elif nota < 9:
    calificacion = "Muy bueno"
else:
    calificacion = "Sobresaliente"

print("Nota:", nota, "→", calificacion)
````

#### Ejemplo aplicado: categoría de edad

````python
edad: int = int(input("Ingresá tu edad: "))

if edad < 13:
    categoria: str = "Niño/a"
elif edad < 18:
    categoria = "Adolescente"
elif edad < 65:
    categoria = "Adulto/a"
else:
    categoria = "Adulto/a mayor"

print("Categoría:", categoria)
````

### **Condicionales anidados**

Es posible colocar un `if` dentro de otro `if`. Esto se llama **anidamiento**. Cada nivel agrega un nivel de indentación.

````python
# Sistema de login: primero verificamos usuario, luego contraseña
user_real: str = "marcelo"
pass_real: str = "1234"

usuario_ingresado: str = input("Usuario: ")

if usuario_ingresado == user_real:
    pass_ingresada: str = input("Contraseña: ")
    if pass_ingresada == pass_real:
        print("✓ Acceso concedido")
    else:
        print("✗ Contraseña incorrecta")
else:
    print("✗ Usuario no encontrado")
````

````python
# Ejemplo: descuento según tipo de cliente y monto de compra
es_cliente_vip: bool = True
monto_compra: float = 1500.0

if es_cliente_vip:
    if monto_compra >= 1000:
        descuento: float = 0.20   # 20%
        print("Cliente VIP con compra alta: 20% de descuento")
    else:
        descuento = 0.10          # 10%
        print("Cliente VIP: 10% de descuento")
else:
    if monto_compra >= 1000:
        descuento = 0.05          # 5%
        print("Compra alta: 5% de descuento")
    else:
        descuento = 0.0
        print("Sin descuento")

precio_final: float = monto_compra * (1 - descuento)
print("Precio final: $", precio_final)
````

### **Expresión ternaria (inline if)**

Python permite escribir un `if/else` simple en **una sola línea**. Es útil cuando queremos asignar un valor según una condición.

```python
variable = valor_si_true  if  condicion  else  valor_si_false
```

````python
# Forma tradicional
nota: float = 7.0

if nota >= 6:
    resultado: str = "Aprobado"
else:
    resultado = "Desaprobado"

print(resultado)
````

````python
# Equivalente con expresión ternaria (una sola línea)
nota: float = 7.0

resultado: str = "Aprobado" if nota >= 6 else "Desaprobado"
print(resultado)
````

````python
# Más ejemplos de expresión ternaria
edad: int = 20
es_mayor: str = "mayor de edad" if edad >= 18 else "menor de edad"
print(es_mayor)

numero: int = -5
signo: str = "positivo" if numero > 0 else ("negativo" if numero < 0 else "cero")
print(signo)
````

### **`match / case` (Python 3.10+)**

A partir de Python 3.10 existe el **structural pattern matching**, similar al `switch/case` de otros lenguajes.
Es útil cuando queremos comparar una variable contra múltiples valores concretos.

```python
match variable:
    case valor_1:
        ...
    case valor_2:
        ...
    case _:          # _ es el caso por defecto (equivalente al else)
        ...
```

> ⚠️ El caso por defecto se escribe con **`_`** (guión bajo), no con la palabra `default`.

````python
# Ejemplo básico: día de la semana
dia: int = 3

match dia:
    case 1:
        print("Lunes")
    case 2:
        print("Martes")
    case 3:
        print("Miércoles")
    case 4:
        print("Jueves")
    case 5:
        print("Viernes")
    case 6:
        print("Sábado")
    case 7:
        print("Domingo")
    case _:
        print("Número de día inválido")
````

````python
# Ejemplo básico: día de la semana
dia: int = 7

match dia:
    case 1:
        print("Lunes")
    case 2:
        print("Martes")
    case 3:
        print("Miércoles")
    case 4:
        print("Jueves")
    case 5:
        print("Viernes")
    case 6 | 7:
        print("Fin de semana!")
    case _:
        print("Número de día inválido")
````

````python
# Ejemplo aplicado: menú de opciones
opcion: str = input("Seleccioná una opción (A/B/C): ").upper()

match opcion:
    case "A":
        print("Elegiste la opción A: Ver reporte")
    case "B":
        print("Elegiste la opción B: Cargar datos")
    case "C":
        print("Elegiste la opción C: Salir")
    case _:
        print("Opción no reconocida")
````

---
### **Resumen: ¿qué estructura usar?**

| Situación | Estructura recomendada |
|---|---|
| Una sola condición | `if` / `else` |
| Muchos rangos o condiciones distintas | `if` / `elif` / `else` |
| Asignar un valor según una condición simple | Expresión ternaria |
| Comparar una variable contra valores concretos | `match / case` |
| Condición dentro de otra condición | Condicionales anidados |
