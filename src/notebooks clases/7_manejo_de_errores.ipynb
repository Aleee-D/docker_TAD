### **¿Qué es una excepción?**

Cuando Python encuentra un problema durante la ejecución de un programa, lanza una **excepción**: una señal de que algo salió mal.

Hay dos tipos de errores:

| Tipo | ¿Cuándo ocurre? | Ejemplo |
|---|---|---|
| **Error de sintaxis** | Antes de ejecutar (el código está mal escrito) | `if x = 5:` |
| **Excepción** | Durante la ejecución (el código está bien escrito pero algo falla) | División por cero, archivo no encontrado |

Si una excepción no se maneja, el programa **se interrumpe**. El manejo de errores nos permite **anticiparnos** a estos problemas y decidir qué hacer cuando ocurren.

En esta notebook veremos:
- Tipos de excepciones más comunes
- `try / except` — capturar errores
- `else` y `finally`
- Múltiples `except`
- `raise` — lanzar excepciones propias
- Excepciones en funciones y en clases

### **`try / except`: capturar excepciones**

La estructura `try / except` nos permite **intentar** ejecutar un bloque de código y, si ocurre una excepción, **capturarla** y decidir qué hacer en lugar de interrumpir el programa.

```python
try:
    # código que puede fallar
except TipoDeExcepcion:
    # qué hacer si ocurre ese error
```

````python
# Sin manejo de errores: el programa se interrumpe
# numero = int(input("Ingresá un número: "))
# print("El doble es:", numero * 2)

# Con manejo de errores: el programa sigue
entrada: str = input("Ingresá un número: ")

try:
    numero: int = int(entrada)
    print("El doble es:", numero * 2)
except ValueError:
    print("Error: eso no es un número válido")
````

````python
# Capturar el mensaje de la excepción con 'as'
try:
    resultado: float = 10 / 0
except ZeroDivisionError as e:
    print("Ocurrió un error:", e)
    print("Tipo:", type(e).__name__)
````

#### Ejemplo aplicado: convertir entrada del usuario de forma segura

````python
def pedir_entero(mensaje: str) -> int:
    """Pide un entero al usuario y repite hasta que ingrese uno válido."""
    while True:
        entrada: str = input(mensaje)
        try:
            return int(entrada)
        except ValueError:
            print("Eso no es un número entero. Intentá de nuevo.")

edad: int = pedir_entero("Ingresá tu edad: ")
print("Edad ingresada:", edad)
````

### **`else` y `finally`**

La estructura completa de manejo de errores tiene cuatro bloques:

```python
try:
    # código que puede fallar
except TipoDeError:
    # se ejecuta SI ocurrió el error
else:
    # se ejecuta SOLO SI no ocurrió ningún error
finally:
    # se ejecuta SIEMPRE, haya error o no
```

| Bloque | ¿Cuándo se ejecuta? |
|---|---|
| `try` | Siempre (es el código a intentar) |
| `except` | Solo si ocurrió una excepción |
| `else` | Solo si NO ocurrió ninguna excepción |
| `finally` | Siempre, pase lo que pase |

````python
def dividir(a: float, b: float) -> None:
    try:
        resultado: float = a / b
    except ZeroDivisionError:
        print("Error: no se puede dividir por cero")
    else:
        # Solo llega acá si NO hubo error
        print("Resultado:", resultado)
    finally:
        # Siempre se ejecuta
        print("--- operación finalizada ---")

dividir(10, 2)
print()
dividir(10, 0)
````

> `finally` es especialmente útil para **liberar recursos**: cerrar archivos, cerrar conexiones a bases de datos, etc., sin importar si hubo error o no. Cuando veamos archivos en la próxima notebook, veremos esto en acción.

### **Múltiples `except`: capturar distintos tipos de error**

Un mismo bloque `try` puede tener varios `except`, uno por cada tipo de error que queramos manejar. Python los evalúa en orden y ejecuta el primero que coincida.

````python
def acceder_elemento(lista: list, indice: str) -> None:
    """Intenta acceder a un elemento de una lista dado un índice como string."""
    try:
        idx: int     = int(indice)        # puede fallar si indice no es numérico
        valor        = lista[idx]         # puede fallar si el índice está fuera de rango
        print("Elemento:", valor)
    except ValueError:
        print("Error: el índice debe ser un número entero")
    except IndexError:
        print("Error: índice", indice, "fuera de rango (lista tiene", len(lista), "elementos)")

mi_lista: list = [10, 20, 30, 40, 50]

acceder_elemento(mi_lista, "2")     # OK
acceder_elemento(mi_lista, "hola")  # ValueError
acceder_elemento(mi_lista, "99")    # IndexError
````

````python
# También podemos capturar varios tipos en un mismo except con una tupla
def procesar_dato(dato) -> float:
    """Intenta convertir un dato a float y calcular su raíz cuadrada."""
    try:
        valor: float = float(dato)
        if valor < 0:
            raise ValueError("No se puede calcular la raíz de un número negativo")
        return valor ** 0.5
    except (ValueError, TypeError) as e:
        print("Error al procesar dato:", e)
        return 0.0

print(procesar_dato(16))      # 4.0
print(procesar_dato("9"))     # 3.0 (convierte el string)
print(procesar_dato("hola"))  # ValueError
print(procesar_dato(-4))      # ValueError personalizado
print(procesar_dato(None))    # TypeError
````

#### `Exception`: capturar cualquier excepción

Si no sabemos exactamente qué error puede ocurrir, podemos usar `Exception` como comodín. Sin embargo, es mejor ser específicos cuando sea posible: capturar todo puede ocultar errores inesperados.

````python
try:
    resultado = 10 / int(input("Divisor: "))
    print("Resultado:", resultado)
except Exception as e:
    print("Ocurrió un error inesperado:")
    print(" Tipo:",    type(e).__name__)
    print(" Mensaje:", e)
````

### **`raise`: lanzar excepciones propias**

`raise` nos permite **lanzar una excepción manualmente** cuando detectamos una situación inválida. Es la forma de comunicar que algo salió mal desde dentro de nuestra función o clase.

```python
raise TipoDeExcepcion("mensaje descriptivo")
```

Esto es fundamental para **validar datos de entrada**: si algo no cumple las condiciones esperadas, lanzamos una excepción con un mensaje claro.

````python
def calcular_promedio(notas: list[float]) -> float:
    """Calcula el promedio de una lista de notas.

    Raises:
        ValueError: si la lista está vacía o alguna nota está fuera del rango 0-10
        TypeError:  si algún elemento no es numérico
    """
    if len(notas) == 0:
        raise ValueError("La lista de notas no puede estar vacía")

    for nota in notas:
        if not isinstance(nota, (int, float)):
            raise TypeError("Todas las notas deben ser numéricas. Valor inválido: " + str(nota))
        if nota < 0 or nota > 10:
            raise ValueError("Las notas deben estar entre 0 y 10. Valor inválido: " + str(nota))

    return sum(notas) / len(notas)


# Casos válidos
print(calcular_promedio([7.0, 8.5, 9.0]))

# Casos inválidos
try:
    print(calcular_promedio([]))
except ValueError as e:
    print("ValueError:", e)

try:
    print(calcular_promedio([7.0, "ocho", 9.0]))
except TypeError as e:
    print("TypeError:", e)

try:
    print(calcular_promedio([7.0, 15.0, 9.0]))
except ValueError as e:
    print("ValueError:", e)
````

#### `raise` para re-lanzar una excepción capturada

````python
# A veces queremos registrar el error pero dejarlo propagarse
def cargar_datos(valor: str) -> int:
    try:
        return int(valor)
    except ValueError as e:
        print("[LOG] Error al cargar dato:", e)
        raise   # re-lanza la misma excepción

try:
    dato: int = cargar_datos("abc")
except ValueError:
    print("El dato no pudo cargarse")
````

### **Excepciones en funciones**

Una buena práctica es **validar los parámetros** al inicio de una función y lanzar excepciones con mensajes descriptivos. Así quien use la función sabe exactamente qué hizo mal.

Los type hints nos dicen qué tipo se espera, y el `raise` nos dice qué valores son válidos.

````python
def calcular_descuento(precio: float, porcentaje: float) -> float:
    """Aplica un descuento a un precio.

    Args:
        precio:     precio original, debe ser mayor a 0
        porcentaje: descuento entre 0 y 100

    Returns:
        Precio con el descuento aplicado.

    Raises:
        ValueError: si precio <= 0 o porcentaje fuera del rango [0, 100]
    """
    if precio <= 0:
        raise ValueError("El precio debe ser mayor a cero. Recibido: " + str(precio))
    if not (0 <= porcentaje <= 100):
        raise ValueError("El porcentaje debe estar entre 0 y 100. Recibido: " + str(porcentaje))

    return precio * (1 - porcentaje / 100)


# Usos válidos
print(calcular_descuento(1000, 20))    # 800.0
print(calcular_descuento(500,   0))    # 500.0

# Usos inválidos
for precio, pct in [(-100, 10), (500, 150), (0, 20)]:
    try:
        print(calcular_descuento(precio, pct))
    except ValueError as e:
        print("ValueError:", e)
````

#### Combinar validación, `raise` y `try/except` en un flujo completo

````python
def buscar_en_diccionario(datos: dict, clave: str) -> str:
    """Busca una clave en un diccionario de forma segura."""
    if not isinstance(datos, dict):
        raise TypeError("El primer argumento debe ser un diccionario")
    if clave not in datos:
        raise KeyError("La clave '" + clave + "' no existe")
    return str(datos[clave])


alumno: dict = {"nombre": "Ana", "legajo": 1001, "nota": 8.5}

for campo in ["nombre", "nota", "telefono"]:
    try:
        valor: str = buscar_en_diccionario(alumno, campo)
        print(campo, "→", valor)
    except KeyError as e:
        print("Campo no encontrado:", e)
````

### **Excepciones en clases**

En las clases que vimos (POO Partes 1 y 2), algunos métodos recibían datos del usuario. Con `raise` podemos proteger esos métodos y garantizar que el objeto siempre esté en un estado válido.

Retomamos `CuentaBancaria` y `Alumno` con validaciones completas.

````python
class CuentaBancaria:
    """Cuenta bancaria con saldo encapsulado y validación de operaciones."""

    def __init__(self, titular: str, saldo_inicial: float = 0.0):
        if not isinstance(titular, str) or len(titular.strip()) == 0:
            raise ValueError("El titular debe ser un texto no vacío")
        if saldo_inicial < 0:
            raise ValueError("El saldo inicial no puede ser negativo")
        self.titular: str    = titular.strip()
        self.__saldo: float  = saldo_inicial

    def depositar(self, monto: float) -> None:
        """Deposita un monto positivo en la cuenta."""
        if not isinstance(monto, (int, float)):
            raise TypeError("El monto debe ser numérico")
        if monto <= 0:
            raise ValueError("El monto a depositar debe ser mayor a cero")
        self.__saldo += monto
        print(f"Depósito de ${monto:.2f} → Saldo: ${self.__saldo:.2f}")

    def retirar(self, monto: float) -> None:
        """Retira un monto si hay saldo suficiente."""
        if not isinstance(monto, (int, float)):
            raise TypeError("El monto debe ser numérico")
        if monto <= 0:
            raise ValueError("El monto a retirar debe ser mayor a cero")
        if monto > self.__saldo:
            raise ValueError(
                f"Saldo insuficiente. Saldo actual: ${self.__saldo:.2f}, "
                f"monto solicitado: ${monto:.2f}"
            )
        self.__saldo -= monto
        print(f"Retiro de ${monto:.2f} → Saldo: ${self.__saldo:.2f}")

    def get_saldo(self) -> float:
        return self.__saldo

    def __str__(self) -> str:
        return f"Cuenta de {self.titular} | Saldo: ${self.__saldo:.2f}"


# Creación válida
cuenta = CuentaBancaria("Ana García", 1000.0)
print(cuenta)

# Operaciones válidas
cuenta.depositar(500.0)
cuenta.retirar(200.0)

# Operaciones inválidas: capturamos los errores
operaciones_invalidas = [
    lambda: cuenta.retirar(5000.0),
    lambda: cuenta.depositar(-100.0),
    lambda: cuenta.depositar("mucho"),
]

print("\n--- Operaciones inválidas ---")
for operacion in operaciones_invalidas:
    try:
        operacion()
    except (ValueError, TypeError) as e:
        print(type(e).__name__ + ":", e)
````

````python
# Creación inválida también se puede capturar
casos_invalidos = [
    ("",    1000.0),    # titular vacío
    ("Ana", -500.0),   # saldo negativo
    (123,   1000.0),   # titular no es string
]

for titular, saldo in casos_invalidos:
    try:
        c = CuentaBancaria(titular, saldo)
    except (ValueError, TypeError) as e:
        print(type(e).__name__ + ":", e)
````

### **Excepciones personalizadas**

Para proyectos más grandes es útil crear **nuestras propias excepciones**, heredando de `Exception`. Esto permite identificar errores específicos de nuestro dominio y separarlos de los errores genéricos de Python.

```python
class MiExcepcion(Exception):
    pass
```

````python
# Excepciones personalizadas para un sistema bancario
class SaldoInsuficienteError(Exception):
    """Se lanza cuando se intenta retirar más de lo disponible."""
    pass

class MontoInvalidoError(Exception):
    """Se lanza cuando el monto de una operación no es válido."""
    pass


class CuentaPremium:
    """Cuenta bancaria con excepciones personalizadas."""

    def __init__(self, titular: str, saldo: float = 0.0):
        self.titular: str   = titular
        self.__saldo: float = saldo

    def retirar(self, monto: float) -> None:
        if monto <= 0:
            raise MontoInvalidoError(f"El monto debe ser positivo. Recibido: {monto}")
        if monto > self.__saldo:
            raise SaldoInsuficienteError(
                f"Saldo insuficiente. Disponible: ${self.__saldo:.2f}"
            )
        self.__saldo -= monto
        print(f"Retiro exitoso. Saldo restante: ${self.__saldo:.2f}")


cuenta = CuentaPremium("Bruno", 500.0)

# Podemos capturar cada tipo de error por separado
try:
    cuenta.retirar(1000.0)
except SaldoInsuficienteError as e:
    print("Sin fondos:", e)
except MontoInvalidoError as e:
    print("Monto inválido:", e)

try:
    cuenta.retirar(-50.0)
except SaldoInsuficienteError as e:
    print("Sin fondos:", e)
except MontoInvalidoError as e:
    print("Monto inválido:", e)
````

### **Resumen**

| Elemento | Descripción |
|---|---|
| `try` | Bloque de código que puede generar una excepción |
| `except TipoError` | Se ejecuta si ocurrió ese error específico |
| `except TipoError as e` | Igual, pero guarda el error en la variable `e` |
| `except (A, B)` | Captura varios tipos en un mismo bloque |
| `except Exception` | Captura cualquier excepción (usar con cuidado) |
| `else` | Se ejecuta solo si NO hubo excepción |
| `finally` | Se ejecuta siempre, haya o no excepción |
| `raise TipoError("msg")` | Lanza una excepción manualmente |
| `raise` | Re-lanza la excepción actual |
| `class MiError(Exception)` | Define una excepción personalizada |

**Excepciones más comunes:**

| Excepción | Causa |
|---|---|
| `ValueError` | Valor de tipo correcto pero inválido |
| `TypeError` | Tipo de dato incorrecto |
| `ZeroDivisionError` | División por cero |
| `IndexError` | Índice fuera de rango |
| `KeyError` | Clave inexistente en diccionario |
| `FileNotFoundError` | Archivo no encontrado |
| `AttributeError` | Atributo o método inexistente |
| `NameError` | Variable no definida |

````python
# Esqueleto completo de manejo de errores
try:
    # código que puede fallar
    pass
except ValueError as e:
    # manejar ValueError
    pass
except (TypeError, KeyError) as e:
    # manejar TypeError o KeyError
    pass
except Exception as e:
    # cualquier otro error no previsto
    pass
else:
    # solo si no hubo errores
    pass
finally:
    # siempre
    pass
````
