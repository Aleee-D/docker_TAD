# Widgets Interactivos en Jupyter

`ipywidgets` permite agregar controles interactivos (sliders, dropdowns, checkboxes) a las notebooks. En lugar de cambiar el código y re-ejecutar, el usuario interactúa directamente con el gráfico o el resultado.

**Casos de uso típicos:**
- Explorar cómo un parámetro afecta un modelo
- Filtrar datos sin escribir código
- Crear herramientas de visualización para otros usuarios

```python
%pip install ipywidgets

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, interactive, Output
from IPython.display import display
```

---
### 1 — `interact` simple

```python
import ipywidgets as widgets
widgets.interact(lambda x: x, x=5)
```

---
`interact` es la forma más simple de crear un widget. Toma una función y genera automáticamente los controles según el tipo de cada parámetro.

#### Ejemplo 1a

El tipo del valor por defecto determina qué widget se crea: un entero genera un `IntSlider`.

```python
# interact llama a la función cada vez que el slider cambia de valor
# la función recibe el valor actual del slider como parámetro
def cuadrado(n):
    return f"{n}² = {n**2}"

# IntSlider define el rango y comportamiento del control
# continuous_update=False: solo ejecuta al soltar el mouse, no mientras se arrastra

deslizador=widgets.IntSlider(value=5,  # valor inicial
                                        min=1,    # mínimo
                                        max=10,   # máximo
                                        step=1,   # incremento
                                        continuous_update=False)

interact(cuadrado, n=deslizador)
```

#### Ejemplo 1b

Un float genera un `FloatSlider`. Acá controlamos la frecuencia de una onda senoidal.

```python
x = np.linspace(0, 5 * np.pi, 500)

def graficar_seno(frecuencia):
    y = np.sin(frecuencia * x)
    plt.figure(figsize=(20, 3))
    plt.plot(x, y, color='steelblue')
    plt.ylim(-1.2, 1.2)
    plt.title(f"y = sin({frecuencia:.1f} · x)")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.show()


deslizador_frecuencia = widgets.FloatSlider(value=1.0, min=0.5, max=5.0, step=0.5, continuous_update=False)

interact(graficar_seno, frecuencia=deslizador_frecuencia)
```

### 2 — `interact` con múltiples controles

Podemos combinar distintos tipos de widgets en una misma función. Un `str` o una lista genera un `Dropdown`.

```python
funciones = {'sin': np.sin, 'cos': np.cos, 'tan': np.tan}

def graficar_funcion(funcion, amplitud):
    y = amplitud * funciones[funcion](x)
    plt.figure(figsize=(20, 3))
    plt.plot(x, y, color='crimson')
    plt.ylim(-3.5, 3.5)
    plt.title(f"y = {amplitud:.1f} · {funcion}(x)")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.show()

# Dropdown: genera un menú desplegable con las opciones indicadas
# FloatSlider: control deslizante para valores decimales
# continuous_update=False: solo redibuja al soltar el mouse

desplegable_funcion = widgets.Dropdown(options=['sin', 'cos'])
deslizable_funcion = widgets.FloatSlider(value=1.0,  # valor inicial
                                         min=0.5,    # mínimo
                                         max=3.0,    # máximo
                                         step=0.5,   # incremento
                                        continuous_update=False) 


interact(graficar_funcion,  funcion=desplegable_funcion, amplitud= deslizable_funcion)
```

###3 — `interactive` (guardar el widget)

`interactive` funciona igual que `interact`, pero devuelve el widget como objeto. Esto permite mostrarlo en otro momento, o acceder a los valores actuales desde otra celda.

### Distribución normal con µ y σ

```python
np.random.seed(42)
# Generamos 500 muestras de una normal estándar — son los datos "observados"
datos = np.random.normal(loc=0, scale=1, size=500)

def graficar_normal(mu, sigma):
    x_curva = np.linspace(-6, 6, 300)
    y_curva = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x_curva - mu) / sigma) ** 2)

    plt.figure(figsize=(8, 4))
    plt.plot(x_curva, y_curva, color='crimson', linewidth=2)
    plt.xlim(-6, 6)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.ylim(0, 0.9)
    plt.title(f"N(µ={mu}, σ={sigma})")
    plt.show()

# interactive es igual a interact pero devuelve el widget como objeto
# esto permite mostrarlo en otro momento o leer sus valores desde otra celda

desplizable_media=widgets.FloatSlider(value=0.0, min=-3.0, max=3.0, step=0.5, description='µ', continuous_update=False)
desplizable_desvio=widgets.FloatSlider(value=1.0, min=0.5, max=3.0, step=0.5, description='σ', continuous_update=False)

w = interactive(graficar_normal, mu=desplizable_media,  sigma=desplizable_desvio)
display(w)
```

#### Widgets manuales con `observe`

Cuando necesitamos más control, creamos los widgets a mano y los conectamos con `.observe()`. Cada vez que cambia el valor de un widget, se ejecuta una función.

### Calculadora de valor futuro

```python
# Output() es un contenedor donde se muestra el resultado — permite limpiarlo y reescribirlo
salida = Output()

# Orden obligatorio: 1) declarar, 2) observe, 3) display
# Si se cambia el orden, los widgets mostrados no tienen el observe conectado

# FloatSlider para valores decimales
# description_width: 'initial' evita que la etiqueta quede cortada
slider_capital = widgets.FloatSlider(value=10000, min=1000, max=100000, step=1000,  description='Capital ($):', style={'description_width': 'initial'},  continuous_update=False)  # solo ejecuta al soltar

slider_tasa = widgets.FloatSlider(value=5.0, min=1.0, max=20.0, step=0.5, description='Tasa anual (%):', style={'description_width': 'initial'},continuous_update=False)

# IntSlider para valores enteros
slider_años = widgets.IntSlider(value=10, min=1, max=30, step=1, description='Años:', style={'description_width': 'initial'}, continuous_update=False)

# change es el evento que dispara observe — no lo usamos, pero la firma lo requiere
def calcular(change):
    C = slider_capital.value       # leemos el valor actual de cada slider
    r = slider_tasa.value / 100    # convertimos % a decimal
    n = slider_años.value
    VF = C * (1 + r) ** n          # fórmula de valor futuro

    salida.clear_output(wait=True)  # borra el output anterior antes de escribir
    with salida:                    # todo lo que se imprime acá va al Output()
        print(f"Capital inicial:  ${C:,.0f}")
        print(f"Tasa anual:       {r*100:.1f}%")
        print(f"Años:             {n}")
        print(f"─" * 30)
        print(f"Valor futuro:     ${VF:,.0f}")
        print(f"Ganancia:         ${VF - C:,.0f}")


# observe conecta cada slider a la función — se ejecuta cada vez que cambia 'value'
slider_capital.observe(calcular, names='value')
slider_tasa.observe(calcular, names='value')
slider_años.observe(calcular, names='value')

display(slider_capital, slider_tasa, slider_años, salida)
calcular(None)  # ejecutamos una vez para mostrar el resultado inicial


# Opción 2 — equivalente explícito (para entender qué hace with internamente)
# salida.__enter__()
# print(f"Capital inicial:  ${C:,.0f}")
# print(f"Tasa anual:       {r*100:.1f}%")
# print(f"Años:             {n}")
# print(f"─" * 30)
# print(f"Valor futuro:     ${VF:,.0f}")
# print(f"Ganancia:         ${VF - C:,.0f}")
# salida.__exit__(None, None, None)  # requiere 3 argumentos: tipo, valor y traceback de error        

# observe conecta
```

#### Tabla interactiva

Dropdown que filtra un DataFrame y actualiza la tabla en tiempo real.

```python
# Dataset de ventas - sintético
np.random.seed(42)
categorias = ['Electrónica', 'Ropa', 'Alimentos', 'Hogar']
sucursales = ['Norte', 'Sur', 'Centro']

df_ventas = pd.DataFrame({
    'sucursal':  np.random.choice(sucursales, 60),   # elige aleatoriamente entre las 3 sucursales
    'categoria': np.random.choice(categorias, 60),   # elige aleatoriamente entre las 4 categorías
    'producto':  [f'Prod-{i:03d}' for i in np.random.randint(1, 50, 60)],  # nombres tipo Prod-007
    'cantidad':  np.random.randint(1, 20, 60),        # entre 1 y 19 unidades
    'precio':    np.random.randint(500, 15000, 60)    # precio unitario en pesos
})
# columna calculada: ingreso total por fila
df_ventas['total'] = df_ventas['cantidad'] * df_ventas['precio']

df_ventas.head()  # muestra las primeras 5 filas para verificar
```

```python
opciones_cat = ['Todas'] + categorias
opciones_suc = ['Todas'] + sucursales
```

Con interact

```python
def filtrar_tabla(categoria='Todas', sucursal='Todas'):
    df_filtrado = df_ventas.copy()

    if categoria != 'Todas':
        df_filtrado = df_filtrado[df_filtrado['categoria'] == categoria]

    if sucursal != 'Todas':
        df_filtrado = df_filtrado[df_filtrado['sucursal'] == sucursal]

    print(f"{len(df_filtrado)} registros  |  Total: ${df_filtrado['total'].sum():,.0f}")
    display(df_filtrado.sort_values('total', ascending=False).head(10))

interact(filtrar_tabla,
         categoria=opciones_cat,
         sucursal=opciones_suc)
```

```python
def filtrar_tabla(categoria, sucursal):
    df_filtrado = df_ventas.copy()

    if categoria != 'Todas':
        df_filtrado = df_filtrado[df_filtrado['categoria'] == categoria]

    if sucursal != 'Todas':
        df_filtrado = df_filtrado[df_filtrado['sucursal'] == sucursal]

    print(f"{len(df_filtrado)} registros  |  Total: ${df_filtrado['total'].sum():,.0f}")
    display(df_filtrado.sort_values('total', ascending=False).head(10))

w = interactive(filtrar_tabla,
                categoria=opciones_cat,
                sucursal=opciones_suc)
display(w)
```

Con observe

```python


drop_cat = widgets.Dropdown(options=opciones_cat, description='Categoría:')
drop_suc = widgets.Dropdown(options=opciones_suc, description='Sucursal:')

salida3 = Output()

def filtrar_tabla(change):
    df_filtrado = df_ventas.copy()

    if drop_cat.value != 'Todas':
        df_filtrado = df_filtrado[df_filtrado['categoria'] == drop_cat.value]

    if drop_suc.value != 'Todas':
        df_filtrado = df_filtrado[df_filtrado['sucursal'] == drop_suc.value]

    salida3.clear_output(wait=True)
    with salida3:
        print(f"{len(df_filtrado)} registros  |  Total: ${df_filtrado['total'].sum():,.0f}")
        display(df_filtrado.sort_values('total', ascending=False).head(10))

drop_cat.observe(filtrar_tabla, names='value')
drop_suc.observe(filtrar_tabla, names='value')

display( widgets.HBox([drop_cat, drop_suc]) )
display(salida3)
filtrar_tabla(None)
```
