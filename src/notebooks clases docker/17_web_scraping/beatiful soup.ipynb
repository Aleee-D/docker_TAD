```python
from bs4 import BeautifulSoup
import requests
```

```python
# Versiones
import bs4 # Solo para el chequeo
print("Versión de BeautifulSoup:",bs4.__version__)
print("Versión de requests:", requests.__version__)
```

```python
url_base = 'https://wikipedia.org/'
html_obtenido = requests.get(url_base)

soup = BeautifulSoup(html_obtenido.text, "html")

print(soup.prettify())   #HTML Pretty
```

### Método find()

```python
# Si queremos quedarnos con un tag
# El método "find" busca el primer elemento de la pagina con ese tag

primer_h1 = soup.find('h1')
print(primer_h1)
```

```python
print(primer_h1.text)
```

### Método find_all()

```python
elementos_divs = soup.find_all('div')
print(elementos_divs)
```

```python
elementos_anchors = soup.find_all('a')
#print(elementos_anchors)

for link in soup.find_all('a'):
    print(link.get('href'))
```

```python
print(soup.get_text())
```

### Método find_all con atributos de etiqueta

```python
# Clase
divs = soup.find_all('div', class_ = "other-project-icon")

for div in divs:
  print(div)
  print(" ")
```
